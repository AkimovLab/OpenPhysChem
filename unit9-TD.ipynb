## Problem 1

Mercury oxide dissociates according to:
$2HgO(s)↔2Hg(g)+O_2 (g)$
At 420 ℃, the total pressure of gases is $5.16 \times 10^4$ Pa, at 450 ℃ it is $10.8 \times 10^4$ Pa. Compute the equilibrium constants at these temperatures as well as the molar enthalpy of dissociation of HgO.

### Solution
Let's say that the partial pressure of $O_2$ is x, then the partial pressure of Hg is 2x and the equilibrium constant is 

$K_p = (2x)^2 x = 4 x^3$

Also, the total pressure is P = 2x + x = 3x, so x = P/3

Hence,

$K_p = 4 (P/3)^3 = 4/27 * P^3$

```python
import math

Pa2bar = 1e-5 # 1 bar = 100 kPa = 10^5 Pa, so 1 Pa = 10^-5 bar
P_420 = 51600 * Pa2bar # bar
P_450 = 108100 * Pa2bar # bar

K_420 = (4.0/27.0) * P_420**3
K_450 = (4.0/27.0) * P_450**3

print(F"K_420 = {K_420} K_450 = {K_450}")

# The ethalpy of dissociation is found from the Van't Hoff's isobar equation
# ln(Kp2/Kp1) = dH_r/R * (1/T1 - 1/T2)

T1 = 420.0 + 273.15 # K
T2 = 450.0 + 273.15 # K
R = 8.314 # J/mol*K

dH = R* math.log(K_450/K_420) / ( 1.0/T1 - 1.0/T2)
print(F"dH = {dH} J")
```

## Problem 2

The equilibrium constant for reaction $CO(g)+2H_2 (g)↔CH_3 OH(g)$ is $K_p=6.09 \times 10^{3}$. Compute the total pressure needed to produce methanol with the 90% yield, if $CO$ and $H_2$ are taken in a stoichiometric proportion. Express your answer in bar and atm units. Round up to the whole number. 

### Solution

```python
# CO + 2 H2 = CH3OH
#  1     2  -> 0            initial moles
#  0.1   0.2   0.9  equilibrium composition  sum = 0.1 + 0.2 + 0.9 = 1.2
# mole fractions:
# 0.1/1.2   0.2/1.2    0.9/1.2
x_CO = 0.1/1.2
x_H2 = 0.2/1.2
x_MeOH = 0.9/1.2

K_x = x_MeOH / (x_CO * x_H2**2)

# K_p = K_x * P^(1 - 1 -2) = K_x / P^2

K_p = 6.09e3
P2 = K_x/K_p
P = math.sqrt(P2) 

# 1 bar = 0.986923 atm
atm2bar = 100/(101.325)
bar2atm = 1.0/atm2bar

print(F"P = {P} bar = {P*bar2atm} atm") # bar? 
```
