# 1. Theory

## 1.1. Reaction Gibbs Free Energy
Chemical potential:
$$
\mu_J = \mu_J^{\circ} + RT \ln a_J
$$

Where: 
- $a_J$ - activity of the component J

For the general reaction:
$$
aA + bB \rightleftharpoons cC + dD
$$

The Gibbs free energy change of a reaction is:
$$
\Delta_r G = c \mu_C + d \mu_D - (a \mu_A + b \mu_B) = c \mu_C^{\circ} + d \mu_D^{\circ} - (a \mu_A^{\circ} + b \mu_B^{\circ}) + RT (c \ln a_C + d \ln a_D - a \ln a_A - b \ln a_B ) = \Delta \mu^{\circ} + RT ln (\frac{a_C^c\ a_D^d}{a_A^a\ a_B^b}) 
$$

or

The Gibbs free energy change of a reaction is:
$$
\Delta_r G = \Delta_r G^{\circ} + RT \ln Q
$$

where:
**standard reaction Gibbs free energy (for 1 mole of reaction)**:
$$
\Delta_r G^{\circ} = \Delta \mu^{\circ}
$$
and
The **reaction quotient** is:
$$
Q = \frac{a_C^c a_D^d}{a_A^a a_B^b}
$$

## 1.2. At equilibrium
$$
\Delta_r G = 0
$$

Thus:
$$
\Delta_r G^\circ = - RT \ln Q{\text{eq}} = - RT \ln K
$$

### A. Mass action law
> At equilibrium, the reaction quotient stays constant
$$
Q{\text{eq}} = K
$$

### B. Equilibrium Constants (K)

In general, defined **in temrs of activities**:
$$
K = \frac{a_C^c a_D^d}{a_A^a a_B^b}
$$

The **equilibrium constant (concentrations, in mol/L)** is:
$$
K_c = \frac{[C]^c [D]^d}{[A]^a [B]^b}
$$

**In terms of pressure (bar units!)**:
$$
K_p = \frac{(p_C)^c (p_D)^d}{(p_A)^a (p_B)^b}
$$

**In terms of mole fractions**:
$$
K_x = \frac{(x_C)^c (x_D)^d}{(x_A)^a (x_B)^b}
$$


For ideal gases: 
$$
P_i = [i]\ R\ T = x_i\ P
$$

so:

**Relation between $K_p$, $K_c$, and $K_x$:**
$$
K_p = K_c (RT)^{\Delta n} = K_x P^{\Delta n}
$$

where  
$$
\Delta n = (c+d) - (a+b)
$$

- $K \gg 1$: products favored  
- $K \ll 1$: reactants favored  
- $K \approx 1$: comparable amounts


## 1.3. Away of Equilibrium

- If $Q < K$: reaction proceeds forward (toward products).  
- If $Q > K$: reaction proceeds backward (toward reactants).  
- If $Q = K$: system is at equilibrium.  

The general equation shall be used:

$$
\Delta_r G = \Delta \mu^{\circ} + RT \ln Q = -RT \ln K + RT \ln Q
$$



## 1.2. Dependence of equilibrium constant of external factors - van’t Hoff’s Equations

The effect of **temperature** on the equilibrium constant

### A. Isobaric equilibrium (van't Hoff's isobar):
$$
\frac{d \ln K_p}{dT} = \frac{\Delta_r H^\circ}{R T^2}
$$

**Integrated form:**
$$
\ln \frac{K_p(T_2)}{K_p(T_1)} = \frac{\Delta_r H^\circ}{R} \left(\frac{1}{T_1} - \frac{1}{T_2}\right)
$$

- If $\Delta_r H^\circ > 0$ (endothermic): $K_p$ increases with $T$  
- If $\Delta_r H^\circ < 0$ (exothermic): $K_p$ decreases with $T$  


### B. Isochoric equilibrium (van't Hoff's isochore):
$$
\frac{d \ln K_c}{dT} = \frac{\Delta_r U^\circ}{R T^2}
$$

**Integrated form:**
$$
\ln \frac{K_c(T_2)}{K_c(T_1)} = \frac{\Delta_r U^\circ}{R} \left(\frac{1}{T_1} - \frac{1}{T_2}\right)
$$

- If $\Delta_r U^\circ > 0$ (endothermic): $K_c$ increases with $T$  
- If $\Delta_r U^\circ < 0$ (exothermic): $K_c$ decreases with $T$  


## 3. Le Châtelier’s Principle

**Statement:**  
*A system at equilibrium, when subjected to a disturbance (concentration, pressure, temperature), shifts in the direction that counteracts the change.*

### Applications:
1. **Concentration**  
   - Add reactant → shifts toward products  
   - Remove product → shifts toward products  

2. **Pressure (gases)**  
   - Increase pressure → shifts toward fewer moles of gas  
   - Decrease pressure → shifts toward more moles of gas  
   - If \(\Delta n = 0\), no effect  

3. **Temperature**  
   - Increase \(T\) → favors endothermic direction  
   - Decrease \(T\) → favors exothermic direction  

4. **Catalyst**  
   - No shift in equilibrium position  
   - Only speeds up attainment of equilibrium

## Example 1

Mercury oxide dissociates according to:
$2HgO(s)↔2Hg(g)+O_2 (g)$
At 420 ℃, the total pressure of gases is $5.16 \times 10^4$ Pa, at 450 ℃ it is $10.8 \times 10^4$ Pa. Compute the equilibrium constants at these temperatures as well as the molar enthalpy of dissociation of HgO.

### Solution
Let's say that the partial pressure of $O_2$ is x, then the partial pressure of Hg is 2x and the equilibrium constant is 

$K_p = (2x)^2 x = 4 x^3$

Also, the total pressure is P = 2x + x = 3x, so x = P/3

Hence,

$K_p = 4 (P/3)^3 = 4/27 * P^3$

```python
import math

Pa2bar = 1e-5 # 1 bar = 100 kPa = 10^5 Pa, so 1 Pa = 10^-5 bar
P_420 = 51600 * Pa2bar # bar
P_450 = 108100 * Pa2bar # bar

K_420 = (4.0/27.0) * P_420**3
K_450 = (4.0/27.0) * P_450**3

print(F"K_420 = {K_420} K_450 = {K_450}")

# The ethalpy of dissociation is found from the Van't Hoff's isobar equation
# ln(Kp2/Kp1) = dH_r/R * (1/T1 - 1/T2)

T1 = 420.0 + 273.15 # K
T2 = 450.0 + 273.15 # K
R = 8.314 # J/mol*K

dH = R* math.log(K_450/K_420) / ( 1.0/T1 - 1.0/T2)
print(F"dH = {dH} J")
```

## Example 2

The equilibrium constant for reaction $CO(g)+2H_2 (g)↔CH_3 OH(g)$ is $K_p=6.09 \times 10^{3}$. Compute the total pressure needed to produce methanol with the 90% yield, if $CO$ and $H_2$ are taken in a stoichiometric proportion. Express your answer in bar and atm units. Round up to the whole number. 

### Solution

```python
# CO(g) + 2 H2(g) = CH3OH(g)
#  1         2    -> 0            initial moles
#  0.1      0.2      0.9     equilibrium composition  sum = 0.1 + 0.2 + 0.9 = 1.2
# mole fractions:
# 0.1/1.2   0.2/1.2    0.9/1.2

x_CO = 0.1/1.2
x_H2 = 0.2/1.2
x_MeOH = 0.9/1.2

K_x = x_MeOH / (x_CO * x_H2**2)

# K_p = K_x * P^(1 - 1 -2) = K_x / P^2

K_p = 6.09e3
P2 = K_x/K_p
P = math.sqrt(P2) 

# 1 bar = 0.986923 atm
atm2bar = 100/(101.325)
bar2atm = 1.0/atm2bar

print(F"P = {P} bar = {P*bar2atm} atm") # bar? 
```

## Example 3 [Eremin Example 9-4]

For reaction 
$$
N_2(g) + 3H_2(g) = 2NH_3(g)
$$
the equilibrium constant at T = 298 K is $K_p = 6.0\times10^5$. Given that $\Delta_f H_{298}^{\circ}(NH_3(g)) = -46.1 kJ\cdot mol^{-1}$, compute the equilibrium constant at T = 500 K.

### Solution

```python
import math 

dH_f_NH3 = -46.1 # kJ/mol

# Convert to SI
dH_f_NH3 = 1000 * dH_f_NH3 # J/mol

# Standard enthalpy of reaction
dH_r = 2 * dH_f_NH3 - 3* 0 - 1*0 

R = 8.314 # J/(mol*K)
T1 = 298 # K
K1 = 6.0e+5 # at T1 

T2 = 500 # K
lnK1 = math.log(K1)

lnK2 = lnK1 + (dH_r/R) * (1/T1 - 1/T2)

K2 = math.exp(lnK2)

print(F"K2 = {K2:.2f}")
```
