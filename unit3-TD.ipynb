# 1. Theory

## 1.1. Introductory definitions

### A. Forms of Energy
- **Kinetic energy (KE, $T$):** energy of motion (translation, rotation, vibration). 
- **Potential energy (PE, $U$):** energy due to position or interactions (chemical bonds, electrostatics). 

### B. Units of Energy and Conversions

| Unit | Symbol | Definition / Relation | Conversion to Joules (J) |
|------|--------|----------------------|--------------------------|
| Joule | J | SI unit of energy | 1 J |
| Kilojoule | kJ | 1000 J | 1 kJ = 1000 J |
| Calorie (thermochemical) | cal | Energy to raise 1 g of water by 1 °C | 1 cal = 4.184 J |
| Kilocalorie (food) | kcal | 1000 cal | 1 kcal = 4184 J |
| Electronvolt | eV | Energy gained by an electron across 1 V | 1 eV = 1.602 × 10⁻¹⁹ J |
| Watt-hour | Wh | Energy delivered by 1 W for 1 hour | 1 Wh = 3600 J |
| Kilowatt-hour | kWh | 1000 Wh | 1 kWh = 3.6 × 10⁶ J |


### C. Modes of Energy Transfer
1. **Work (W):** energy transfer via macroscopic uniform motion against external forces
2. **Heat (Q):** energy transfer due to random motion of particles

### D. Energy balance = Connection of 3 Key Properties
$$
E_{tot} = [ T_{tr,CM} + T_{rot,RB}] + [T_{internal} + U_{int}(\textrm{internal DOF})] + U_{rel}(\textrm{ceters of mass, orientations})
$$

1. **Work** - related to potential energy at the macroscopic level (e.g. cylinder in engine), realized via potential energy of intermolecular interactions that depends on relative positions of molecules with respect to each other, $U_{rel}(\textrm{ceters of mass, orientations})$
2. **Heat** - related to kinetic energy of translational and rotational motion of molecules as rigid bodies, $T_{tr,CM} + T_{rot,RB}$
3. **Internal energy** - kinetic energy of intramolecular motions, $T_{internal}$, as well as the potential energy of intramolecular interactions $U_{int}(\textrm{internal DOF})$


## 1.2. The 1st Law of Thermodynamics  

### A. Formulation

**Integral form:**  
$$
\Delta U = Q + W
$$  
- \($ \Delta U $\): change in internal energy  
- \($ Q $\): heat absorbed by the system  
- \($ W $\): work done **on/by** the system  

**Differential form:**  
$$
dU = \delta Q + \delta W
$$  

**Verbal form:**
> The energy of isolated system is constant. The energy of a non-isolated system can change due to heat transferred to the system from the external environment ( $𝛿𝑄_{𝑒𝑥𝑡}$) and due to work ($𝛿𝑊$) made by the system on the environment. 

**Even simpler verbal form:**
> The perpetuum mobile of the first kind is impossible

- **Perpetuum mobile of the first kind = perpetual motion machine of the first kind** - a device whose parts are not only in permanent motion, but moreover is able to provide useful work without input of external energy (e.g. warmth) and without change of the physical or chemical status of its parts. A perpetual motion mahine of the first kind does not exist. It would contradict the first law of thermodynamics


**In simple words:**
> It is the law of the energy conservation or the impossibility of the perpetuum mobile of the first kind (creates energy from nothing)

### B. Sign Conventions

- **Heat (Q):**  
  - \( Q > 0 \): heat absorbed by system (heating up). The energy of the system increases.
  - \( Q < 0 \): heat released by system (cooling down). The energy of the system decreases.
- **Work (W):**  
  - \( W > 0 \): work done **on** the system (compression). The energy of the system increases.
  - \( W < 0 \): work done **by** the system (expansion). The energy of the system decreases.

## 1.3. Work 

The general expression is:

$$
|dW| = |F| \times |dq| 
$$

- \(dW \) - work
- \(F \)- generalized force
- \(dq \)- generalized coordinate

### A. Generalized Forces and Coordinates

| Generalized Coordinate (extensive) | Generalized Force (intensive)  | Kind of work |
|-------------------------|------------------|------------------------|
| Volume (V)             | Pressure (P)     | Mechanical work: Gas expansion |
| Length (L)             | Force (F)        | Mechanical work: stretching or compression |
| Charge (q)             | Electric potential ($\phi$) | Work of charging | 
| Magnetic induction (B) | Magnetic field (H) | Work of magnetization |
| Surface area ($\Omega$)| Surface tension ($\sigma$) | Work of creating surface |
| Number of moles (n)    | Chemical potential | Work of mixing |


### B. Work of Gas Expansion

<img src="pics/Slide2.PNG" width="50%">

- **Reversible process (ideal gas, isothermal):**  

In this case, $P_{ext} = P_{int}$ The internal pressure is determined by the gas laws. 

$$
|W| = \int_{V_i}^{V_f} P_{ext} \, dV = \int_{V_i}^{V_f} P_{int} \, dV = \int_{V_i}^{V_f} \frac{nRT}{V} \, dV= nRT \ln \frac{V_f}{V_i}
$$  

Here,
- $V_i$ - initial volume of gas
- $V_f$ - final volume of gas

- **Irreversible process (constant external pressure):**  

In this case, $P_{ext}$ is constant, and is determines the final volume of expansion (when internal pressure would become equal to the external one)
$$
W = |W| = \int_{V_i}^{V_f} P_{ext} \, dV = P_{\text{ext}} (V_f - V_i)
$$  


## 1.4. Heat

### A. Heat Capacity
- **Heat capacity:** the amount of heat required to raise temperature by 1 K (extensive property).  
$$
C = \frac{dQ}{dT} = \frac{Q}{\Delta T}
$$  
- At constant volume (isochoric): $$ C_V = \left(\frac{\partial U}{\partial T}\right)_V = \left(\frac{\delta Q}{dT}\right)_V $$.  
- At constant pressure (isobaric): $$ C_P = \left(\frac{\partial H}{\partial T}\right)_P = \left(\frac{\delta Q}{dT}\right)_P $$.  

Here, 
- \(U \) - internal energy
- \(H \) - enthalpy

Additional types:
- **Specific heat capacity** - per unit of mass (intensive property)
$$
C_s = \frac{C}{m}
$$

- **Molar heat capacity** - per number of moles (intensive property)
$$
C_m = \frac{C}{n}
$$

Here
- \( m \) - mass
- \( n \) - number of moles


### B. Exothermic and Endothermic Processes

**Exothermic** - system releases energy as heat (Q < 0)
Examples:
- Combustion: $𝐶_2𝐻_5𝑂𝐻 + 𝑂_2 \rightarrow 𝐶𝑂_2 + 𝐻_2𝑂$
- Reactions: $𝑁𝑎 + 𝐻_2𝑂 \rightarrow 𝑁𝑎𝑂𝐻 + 𝐻_2$
- Condensation: $𝐻_2𝑂(𝑣𝑎𝑝.) \rightarrow 𝐻_2𝑂(liq.)$
- Mixing: $𝐻_2𝑂 + 𝐻_2𝑆𝑂_4(conc.) \rightarrow ... $

**Endothermic** - system absorbs energy via heat (Q > 0)
Examples:
- Photosynthesis: $𝐶𝑂_2 + 𝐻_2O + ℎ𝜈 \rightarrow 𝑐𝑎𝑟𝑏𝑠$
- Melting of ice, evaporation, sublimation
- Cracking of alkanes
- Coldpatch in a first-aid kit: $𝑁𝐻_4𝐶𝑙(s) + H_2O(liq.) \rightarrow NH_4^+(solv.) + Cl^−(solv.)$ same with $𝑁𝐻_4𝑁𝑂_3$

### C. Molecular origins of heat capacity

Heat capacity reflects available degrees of freedom (translation, rotation, vibration).  

> **Equipartition theorem:** each quadratic term in molecular energy adds $\frac{1}{2}𝑘_𝐵\ 𝑇$ to the internal energy per molecule or $\frac{1}{2} 𝑅𝑇$ to isochoric molar internal energy.  

- Monoatomic ideal gas: $$ C_{m,V} = \frac{3}{2} RT$$ 
- Diatomic ideal gas: $$ C_{m,V} = \frac{5}{2} RT$$ 

Because: 
- For monoatomic gas (3 terms): $$E_{kin} = \frac{m v_x^2}{2} + \frac{m v_y^2}{2} + \frac{m v_z^2}{2}$$
- For diatomic gas (5 terms): $$E_{kin} = \frac{m v_x^2}{2} + \frac{m v_y^2}{2} + \frac{m v_z^2}{2} + \frac{I_{xx}\omega_x^2}{2} + \frac{I_{yy}\omega_y^2}{2}$$


## 1.5. Internal Energy

**Internal energy (U)** - sum of all kinetic and potential energy contributions due to atoms, electrons, spin, etc.

Properties:
- useful at constant volume (e.g. bomb calorimetry)
- it is an extensive property;
- depends on temperature and pressure, $𝑈 = 𝑈(𝑇,𝑃,...)$
- absolute value is not important; only its change, $\Delta 𝑈$, matters
- it is a state function ($\Delta 𝑈 = 0$ for any cyclic path)
- $\Delta 𝑈 = 0$ for isothermal expansion of an ideal gas


## 1.6 Enthalpy
- **Enthalpy (H):**  
$$
H = U + PV
$$  

Properties:
- Useful at constant pressure processes (chemistry, biology)
- **Significance:** Change of enthalpy is equal to the thermal effect of reaction under constant pressure

$$
\Delta 𝐻 = \Delta 𝑈 + \Delta (𝑃𝑉) = \Delta 𝑈 + (𝑃_𝑓\ 𝑉_𝑓 - 𝑃_𝑖\ 𝑉_𝑖 ) = \Delta 𝑈 + P\Delta 𝑉 = (Q + W) + P\Delta 𝑉 = Q - P\Delta V + P\Delta 𝑉 = Q_P
$$
Used:
- Constante pressure setup, so $P_i = P_f = P$
- 1st law of TD, so $\Delta U = Q_P + W$
- Here, $Q_P$ is the thermal effect (heat) of the process under constant pressure
- Work done by system (gas) expansion: $W = -P\Delta V$


## 1.7. Computing Changes of TD Functions in Various TD Processes (with W<0 for expansion)

Differential form of the 1st law of TD:
$$
dU = \left( \frac{\partial U}{\partial T}\right)_V dT + \left( \frac{\partial U}{\partial V}\right)_T dV
$$

### A. Isothermal process (T = const)
- **Internal energy:**
$$\left( \frac{\partial U}{\partial T}\right)_V = 0$$
so
$$
dU = \left( \frac{\partial U}{\partial T}\right)_V dT + \left( \frac{\partial U}{\partial V}\right)_T dV = \
\left( \frac{\partial U}{\partial V}\right)_T dV
$$

For the ideal gas, U, doesn't depend on V, so $\left( \frac{\partial U}{\partial V}\right)_T = 0$ ans hence:
$$
dU = \left( \frac{\partial U}{\partial V}\right)_T dV = 0
$$
so:
$\Delta U = 0$ and so $Q = -W$
- **Work:**
We found before that (with the sign convention), $W = -nRT \ln \frac{V_f}{V_i}$
- **Heat:** follows from the above
- **Enthalpy:**
Finally, because T = const, PV = const, that means $\Delta(PV) = 0$, hence $\Delta H = \Delta U + \Delta (PV) = \Delta U = 0$

### B. Isochoric process (V = const)
- **Internal energy**
$$\left( \frac{\partial U}{\partial V}\right)_T = 0$$
so
$$
dU = \left( \frac{\partial U}{\partial T}\right)_V dT + \left( \frac{\partial U}{\partial V}\right)_T dV = \
\left( \frac{\partial U}{\partial T}\right)_V dT = n C_{m,V} dT
$$
or, in integral form,
$$
\Delta U = n \int_{T_i}^{T_f} C_{m, V}(T) dT
$$
for the ideal gas, $C_{m,V}(T) = const = C_{m,V}$, so:
$$
\Delta U = n \int_{T_i}^{T_f} C_{m, V}(T) dT = n C_{m, V} \Delta T
$$
- **Work:** There is no expansion, so $W = 0$
- **Heat:** Follows from the 1st TD law and the above results
- **Enthalpy:** Since V = const, $\Delta (PV) = V \Delta P$, so:
$$
\Delta H = \Delta U + \Delta (PV) = \Delta U + V \Delta P
$$

### C. Isobaric process (P = const)
- **Enthalpy:** - by definition,
$$
C_{m,P} = \frac{1}{n} \left( \frac{\partial Q_P}{\partial T}\right)_P
$$
so
$$
Q_P = n \int_{T_i}^{T_f} C_{m,P} dT = n C_{m,P} \Delta T
$$
- **Internal energy** 
Note that for isobaric process, $\Delta (PV) = P \Delta V$, so 
$$
\Delta U = \Delta H - \Delta (PV) = \Delta H - P \Delta V = n C_{m,P} \Delta T - P \Delta V
$$
- **Work:** - derived directly from definition (constant generalized force)
$$
W = -P \Delta V
$$
- **Heat:** - follows from the 1st law of TD:
$$
Q = \Delta U - W = n C_{m,P} \Delta T - P \Delta V - (-P \Delta V) = n C_{m,P} \Delta T
$$

### D. Adiabatc process (Q = 0)
- **Heat:** - by definition, $Q = 0$
- **Equation of the adiabatic gas expansion:**
Starting with
$$
dU = \left( \frac{\partial U}{\partial T}\right)_V dT + \left( \frac{\partial U}{\partial V}\right)_T dV
$$
Because of $Q = 0$, $dU = \delta W = -P dV$, so:
$$
\left( \frac{\partial U}{\partial T}\right)_V dT + \left( \frac{\partial U}{\partial V}\right)_T dV = -P dV
$$

For the ideal gas, $\left( \frac{\partial U}{\partial V}\right)_T = 0$ and $\left( \frac{\partial U}{\partial T}\right)_V = n C_{m, V} dT$, so we obtain:
$$
nC_{m,V} dT = -P dV = - n \frac{RT}{V} dV
$$
$$
C_{m, V} \frac{dT}{T} = - R \frac{dV}{V}
$$
Integrating both sides, we obtain
$$
C_{m, V} \ln T = - R \ln V
$$
Or 
$$
\boxed{ T V^{\gamma - 1} = const} \quad \textrm{Adiabat equation}
$$ 

where $\gamma = \frac{C_{m,P}}{C_{m,V}}$
- **Work:** use the adiabat equation to compute the work via $W = -\int_{V_i}^{V_f} P dV$, where $P$ should be expressed via adiabat equation. 
- **Internal energy:** - since Q = 0, $\Delta U = W$

### E. Summary of Formulas 

| Process      | Heat (Q) | Work (W, on system) | ΔU | ΔH |
|--------------|----------|---------------------|----|----|
| **Isothermal (T=const, ideal gas)** | $$nRT \ln \frac{V_f}{V_i}$$ | $$-nRT \ln \frac{V_f}{V_i}$$ | 0 | 0 |
| **Isochoric (V=const)** | $$nC_{m,V} \Delta T$$ | 0 | $$nC_{m,V} \Delta T$$ | $$nC_{m,V} \Delta T + V \Delta P$$ |
| **Isobaric (P=const)**  | $$nC_{m,P} \Delta T$$ | $$-P \Delta V$$ | $$nC_{m,P} \Delta T - P \Delta V$$ | $$nC_{m,P} \Delta T$$ |
| **Adiabatic (Q=0)** | 0 | $$nC_{m,V} \Delta T$$ | $$nC_{m,V} \Delta T$$ | $$nC_{m,P} \Delta T$$ |

## Example 1

Compute $\Delta U$, $\Delta H$, q, and w (done by gas) in a reversible isothermal expansion of nitrogen from 0.5 to 4 $m^3$. The initial temperature and pressure of gas are T=26.8℃ and P=93.2 kPa. Report the results in kJ units rounded up to 1 decimal place. 

### Solution

```python
import math

# Constants
mmHg2atm = 1.0/760.0  # P[atm] = P[mmHg] * mmHg2atm
R = 0.0821 # L*atm/mol*K
atm2Pa = 101325.0 # P[Pa] = P[atm] * atm2Pa
cal2J = 4.184 # E[J] = E[cal] * cal2J
J2cal = 1.0/4.184
```

```python
V1 = 0.5 # m^3
V2 = 4.0 # m^3

T1 = 26.8 + 273.15 # K

# Because the expansion is isothermal, 
T2 = T1

# We know the initial pressure of N2, so we can compute the number of moles
# P1 * V1 = n * R *T1
P1 = 93200.0 # Pa
R = 8.314 # J/mol*K
n = P1 * V1 / (R * T1)  # the number of moles

# For isothermal process, we have
dU = 0.0
dH = 0.0 
q = n * R * T1 * math.log(V2/V1)
w = -q

print(F"dU = {dU} J, dH = {dH} J, q = {q} J, w = {w} J")
```

## Example 2

Three moles of an ideal monoatomic gas residing at the starting temperature of $T_1=350$ K and pressure of $P_1=5$ atm expand adiabatically and reversibly to the final pressure of $P_2=1$ atm. Compute final temperature and volume (L), as well as the work (kcal) done by gas, and the change of internal energy (kcal) and enthalpy (kcal) in this process. 

### Solution 

```python
n = 3.0    # mol
P1 = 5.0   # atm
P1 = P1 * atm2Pa # now in Pa
P2 = 1.0   # atm
P2 = P2 * atm2Pa # now in Pa
Cv = 3.0   # J/mol*K
T1 = 350.0 # K 
R = 8.314  # J/mol*K

# First, let's compute what we can:
# initial volume: P1 V1 = n * R * T1
V1 = n * R * T1 / P1
print(F"Initial volume = {V1} m^3, {V1 * 1000.0} L")

# In the adiabatic process, we have the equation of the adiabat
# T * V^{gamma-1} = const
# but also we have the ideal gas EOS PV = nRT, so:
#  PV/(nR) * V^{gamma-1} = const
# or PV^{gamma} = const
# that is P1 * V1^{gamma} = P2 * V2^{gamma}
# that is: P1/P2 = (V2/V1)^{gamma}

# gamma = Cp/Cv, so gamma - 1 = (Cp - Cv)/Cv = R/Cv 
Cv = 1.5 * R
gamma = (Cv + R) / Cv

# now
V2 = V1 * math.pow( (P1/P2), 1.0/gamma )
print(F"Final volume = {V2} m^3, {V2 * 1000.0} L")

# now let's compute the final temperature 
T2 = P2 * V2 / (n * R)
print(F"Final temperature = {T2} K")

# Because it is an adiabatic process, we compute dU as:
dU = n * Cv * (T2-T1)
print(F"dU = {dU} J, {dU * J2cal} cal")

# Because in the adiabatic process, we have w = n * Cv * (T2-T1)
w = n * Cv * (T2-T1)
print(F"w = {w} J, {w * J2cal} cal")

# Also, in the adiabatic process, we have: dH = n * Cp * (T2-T1)
# Cp = Cv + R
Cp = Cv + R
dH = n * Cp * (T2-T1)
print(F"dH = {w} J, {dH * J2cal} cal")
```

## Example 3

Eremin 2.18

One mole of methane taken at 25 Celsius and 1 atm is heated at constant pressure up to doubling its volume. Molar heat capacity of methane is given by the expression:

$C_{p,m} = 5.34 + 0.0115 T$  &nbsp;  $cal\times mol^{-1} K^{-1}$

Calculate $\Delta U$ and $\Delta H$ for this process. Consider methane to be the ideal gas.

### Solution

$\Delta H = \int_{T_1}^{T_2} dH = n \int_{T_1}^{T_2} C_{p, m} dT$

since $C_{p, m} = a + b \times T$

then:

$\Delta H = n [ a \Delta T + \frac{b}{2} (T_2^2 - T_1^2)  ] $

```python
# dH = n * C_p,m * dT  
# integral: 

a = 5.34 # cal / mol * K
b = 0.0115 # cal / mol * K^2

n = 1.0 # mole
T1 = 25 + 273.15 # K
P1 = 1.0 # atm
P1 = P1 * 101325 # Pa
P2 = P1
R = 8.314 # J/mol*K


# compute the initial volume:  PV = nRT
V1 = n*R*T1/P1  # m^3

# The final volume is twice that:
V2 = 2.0*V1

# Because it is constant pressure, the temperature is twice the initial temperature:
T2 = 2.0*T1

dT = T2 - T1

# So, compute:
dH = n * a * dT + 0.5*b*(T2**2 - T1**2)

print(F"dH = {dH} J")

# H = U + PV, so 
dU = dH - (P2*V2 - P1*V1)
print(F"dU = {dU} J")
```
