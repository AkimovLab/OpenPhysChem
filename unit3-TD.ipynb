## Problem 1

Compute $\Delta U$, $\Delta H$, q, and w (done by gas) in a reversible isothermal expansion of nitrogen from 0.5 to 4 $m^3$. The initial temperature and pressure of gas are T=26.8℃ and P=93.2 kPa. Report the results in kJ units rounded up to 1 decimal place. 

### Solution

```python
import math

# Constants
mmHg2atm = 1.0/760.0  # P[atm] = P[mmHg] * mmHg2atm
R = 0.0821 # L*atm/mol*K
atm2Pa = 101325.0 # P[Pa] = P[atm] * atm2Pa
cal2J = 4.184 # E[J] = E[cal] * cal2J
J2cal = 1.0/4.184
```

```python
V1 = 0.5 # m^3
V2 = 4.0 # m^3

T1 = 26.8 + 273.15 # K

# Because the expansion is isothermal, 
T2 = T1

# We know the initial pressure of N2, so we can compute the number of moles
# P1 * V1 = n * R *T1
P1 = 93200.0 # Pa
R = 8.314 # J/mol*K
n = P1 * V1 / (R * T1)  # the number of moles

# For isothermal process, we have
dU = 0.0
dH = 0.0 
q = n * R * T1 * math.log(V2/V1)
w = -q

print(F"dU = {dU} J, dH = {dH} J, q = {q} J, w = {w} J")
```

## Problem 2

Three moles of an ideal monoatomic gas residing at the starting temperature of $T_1=350$ K and pressure of $P_1=5$ atm expand adiabatically and reversibly to the final pressure of $P_2=1$ atm. Compute final temperature and volume (L), as well as the work (kcal) done by gas, and the change of internal energy (kcal) and enthalpy (kcal) in this process. 

### Solution 

```python
n = 3.0    # mol
P1 = 5.0   # atm
P1 = P1 * atm2Pa # now in Pa
P2 = 1.0   # atm
P2 = P2 * atm2Pa # now in Pa
Cv = 3.0   # J/mol*K
T1 = 350.0 # K 
R = 8.314  # J/mol*K

# First, let's compute what we can:
# initial volume: P1 V1 = n * R * T1
V1 = n * R * T1 / P1
print(F"Initial volume = {V1} m^3, {V1 * 1000.0} L")

# In the adiabatic process, we have the equation of the adiabat
# T * V^{gamma-1} = const
# but also we have the ideal gas EOS PV = nRT, so:
#  PV/(nR) * V^{gamma-1} = const
# or PV^{gamma} = const
# that is P1 * V1^{gamma} = P2 * V2^{gamma}
# that is: P1/P2 = (V2/V1)^{gamma}

# gamma = Cp/Cv, so gamma - 1 = (Cp - Cv)/Cv = R/Cv 
Cv = 1.5 * R
gamma = (Cv + R) / Cv

# now
V2 = V1 * math.pow( (P1/P2), 1.0/gamma )
print(F"Final volume = {V2} m^3, {V2 * 1000.0} L")

# now let's compute the final temperature 
T2 = P2 * V2 / (n * R)
print(F"Final temperature = {T2} K")

# Because it is an adiabatic process, we compute dU as:
dU = n * Cv * (T2-T1)
print(F"dU = {dU} J, {dU * J2cal} cal")

# Because in the adiabatic process, we have w = n * Cv * (T2-T1)
w = n * Cv * (T2-T1)
print(F"w = {w} J, {w * J2cal} cal")

# Also, in the adiabatic process, we have: dH = n * Cp * (T2-T1)
# Cp = Cv + R
Cp = Cv + R
dH = n * Cp * (T2-T1)
print(F"dH = {w} J, {dH * J2cal} cal")
```
