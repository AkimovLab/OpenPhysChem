# Unit 7: **Applications of thermodynamics: phase equilibria in 1-component systems. Clapeyron and Clausius-Clapeyron equations.**

# Significance

* **Boiling point elevation & distillation**
– Predict how vapor pressure changes with temperature (important in purification, solvent removal).

* **Phase diagrams of pure substances**
– Understanding melting/freezing points, sublimation, and how pressure shifts equilibrium (e.g., CO₂ dry ice sublimation).

* **Enthalpy of phase transitions**
– Using Clausius–Clapeyron to calculate enthalpy of vaporization or sublimation from vapor pressure data.

* **Material science**
– Determining conditions for solid–liquid–gas equilibria in designing solvents, refrigerants, or supercritical fluids.

* **Cryogenics & low-temperature chemistry**
– Predicting behavior of liquid nitrogen, helium, and other cryogens in lab storage and experimental setups.

* **Sterilization and autoclaves**
– Vapor pressure–temperature relationships explain how increasing pressure raises boiling point of water → achieving sterilization temperatures above 100 °C.

* **Respiratory physiology**
– Water vapor pressure in lungs (Clausius–Clapeyron governs dependence on body temperature), affecting partial pressures of gases for respiration calculations.

* **Drug formulation & storage**
– Predicting sublimation or evaporation rates of solvents/active ingredients (important for freeze-dried pharmaceuticals).

* **Cryopreservation of tissues/cells**
– Understanding phase equilibria of water and cryoprotectants, avoiding ice crystal damage.

* **Anesthetics and inhaled gases**
– Volatile anesthetics depend on accurate vapor pressure data (often derived using Clausius–Clapeyron) for dosing.

* **Body temperature & sweating**
– Relation between vapor pressure of water and temperature explains evaporative cooling efficiency in thermoregulation.
