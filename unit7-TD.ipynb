# 1. Theory

## 1.1. Relevant Terminology

### A. States of Matter

- **Plasma**

  - A charged gas
  - Composed of ions

- **Gas**

  - Interactions weak or negligible
  - Atoms move fast and long
  - A compressible fluid

- **Liquid**

  - Interactions are weaker than in solids
  - No long-range order
  - Atoms vibrate and diffuse notably
  - Nearly incompressible fluid

- **Solid**

  - Strong interactions
  - Long-range order
  - Atoms vibrate

### B. Phases

**Phase** = specific state of matter that is uniform throughout in composition and physical state

- Gas state = only 1 phase
- Liquid state = only 1 substance has 2 liquid phases (He-I and He-II)
- Solid state = many phases!!!

Examples: carbon phases - graphite, diamond, fullerenes, etc.

### C. Phase Transitions
- **First-order**: discontinuous in entropy or volume (fusion, vaporization, sublimation).  
- **Second-order**: continuous in entropy/volume but discontinuous in derivatives (superconductivity, magnetic transitions).


### D. Phase Diagrams
- Solid–Liquid–Gas boundaries given by phase equilibria.  
- Triple Point:
$$
\mu_i^{(s)} = \mu_i^{(l)} = \mu_i^{(g)}
$$

- Critical Point: liquid–vapor distinction disappears.

### E. Gibbs Phase Rule
General formula:
$$
F = C - P + 2
$$

For condensed systems (pressure nearly constant):
$$
F = C - P + 1
$$

where:
- **F**: the number of degrees of freedom (variables like P, V, T, or composition) that can be varied without causing phase transition
- **C**: the number of components (e.g. water, salt, etc.)
- **P**: the number of phases (e.g. liquid, solid, etc.)


## 1.2. Molar Gibbs Energy and Chemical Potential
- Gibbs Free Energy:
$$
G = H - TS = U + PV - TS
$$

- Molar Gibbs Energy:
$$
\bar{G} = \frac{G}{n}
$$

- Chemical Potential of component \( i \):
$$
\mu_i = \left( \frac{\partial G}{\partial n_i} \right)_{T,P,n_{j \neq i}}
$$

## 1.3. Condition of Equilibrium
- For a closed system at constant \( T \) and \( P \):
$$
dG \leq 0
$$

- At equilibrium (phase coexistence):
$$
\mu_i^{(\alpha)} = \mu_i^{(\beta)}
$$


## 1.4. Variation of Chemical Potential with respect to P and T

### A. General Relationships

From the fundamental relation:
$$
d\mu = -S_m \, dT + V_m \, dP
$$

- At constant \( T \):
$$
\left( \frac{\partial \mu}{\partial P} \right)_T = V_m
$$

- At constant \( P \):
$$
\left( \frac{\partial \mu}{\partial T} \right)_P = -S_m
$$


### B. Clapeyron Equation
For a phase transition:
$$
\frac{dP}{dT} = \frac{\Delta S_{trans}}{\Delta V_{trans}}
= \frac{\Delta H_{trans}}{T \, \Delta V_{trans}}
$$

### C. Clausius–Clapeyron Equation
Approximation for vaporization:
$$
\frac{d \ln P}{dT} = \frac{\Delta H_{vap}}{R T^2}
$$

Integrated form:
$$
\ln P = -\frac{\Delta H_{vap}}{R} \cdot \frac{1}{T} + C
$$

### D. Significance and Applicability:

> The significance of the Clapeyron and Clausius-Clapeyron equations - they describe the phase boundary conditions

- **Clapeyron equation**: For liquid-liquid, solid-solid, and liquid-solid equilibria
- **Clausius-Clapeyron equation**: For liquid-vapor equilibrium. Explains the change of boiling T with the pressure change.

- **Normal boiling point**: at 1 atm (760 Torr = 101325 Pa)
- **Standard boiling point**: at 1 bar ($10^5$ Pa)

## Problem 1

The density of liquid and solid tin (Sn) at its melting temperature of 231.9 ℃ are 6.980 $\frac{g}{cm^3}$ and 7.184 $\frac{g}{cm^3}$, respectively. Molar fusion enthalpy of tin is $\Delta_{fus} H_m=1.690 \frac{kcal}{mol}$. Compute melting temperature at pressure of 500 atm. Take the molar mass of tin to be $M_{Sn}=118.7 \frac{g}{mol}$.

### Solution

```python
import math

# Conversion factors
cal2J = 4.184  # 1 cal = cal2J J
atm2Pa = 101325 # 1 atm = atm2Pa Pa

# Conditions
dH_fus = 1600.0 * cal2J # cal/mol -> J/mol
T_fus = 231.9 + 273.15 # K
M_sn = 118.7 # g/mol
rho_liq = 6.980 # g/cm^3
rho_sol = 7.184 # g/cm^3


# Lets compute molar volumes of Sn in liquid and solid phases
# rho = M/V => V = M/rho
V_liq = M_sn/rho_liq
V_sol = M_sn/rho_sol
print(F"Molar volumes: liq = {V_liq} solid = {V_sol}")

dV = V_liq - V_sol # phase transition - melting: sol -> liq, in cm^3/mol

# Convert to SI units: m^3/mol
dV = dV * 1e-6 

# First let's compute the slope of pressure-vs-temperature curve,
# we use Clapeyron's equation:

dPdT = dH_fus/(dV * T_fus)

# Note that dT/dP = (dP/dT)^-1. so:
dTdP = 1.0/dPdT
print(F"dTdP = {dTdP}")


# So, compute the change of the melting temperature vs change of pressure as:
dP = (500 - 1)*atm2Pa # Pa

dT = dTdP * dP

# dT = T_fus(500 atm) - T_fus(1 atm), so:
T_fus_500 = T_fus + dT

print(F"Fusion temperature at 500 atm is {T_fus_500} K, {T_fus_500 - 273.15} Celsius")
```

## Problem 2

The vapor pressure (Torr units) of some substance is described by equation: $ln⁡(p)=16.255-\frac{2501.8}{T}$. Compute molar enthalpy of vaporization (kJ/mol) and normal boiling temperature (in Celsius) of this liquid. 

### Solution

We use the Clausius-Clapeyron equation:

$ln(p) = ln(p') + \frac{\Delta_{vap} H }{R} (\frac{1}{T'} - \frac{1}{T})$

That means that $ \frac{\Delta_{vap} H }{R} = 2501.5$

```python
R = 8.314 # J/mol*K
# So:

dH_vap = 2501.5 * R

print(F"dH_vap = {dH_vap} J")


# Now let's compute the normal boiling point 
P = 1.0 # atm 
atm2torr = 760.0 # 1 atm = atm2torr torr
P = P * atm2torr  # in torr

T = 2501.5/(16.255 - math.log(P))
print(F"Boiling temperature = {T} K = {T-273.15} Celsius")
```
