# Unit 16: **Approximate methods of chemical kinetics: Quasi-equilibrium and steady-state approximations. Deriving rate laws from mechanisms of chemical reactions.**


# 1. Theory

## 1.1. Complex Reaction Mechanisms

Many reactions proceed via several elementary steps:

$$
A \rightleftharpoons I \rightarrow P
$$

where \(I\) is an intermediate.

We use **approximations** to simplify the kinetic equations.


## 1.2. Quasi-Equilibrium Approximation (QEA)

### A. Idea
Early steps are **fast and reversible**, reaching equilibrium before the slow (rate-determining) step.

$$
A \underset{k_{-1}}{\stackrel{k_1}{\rightleftharpoons}} I
\quad \text{(fast)}
$$

$$
I \xrightarrow{k_2} P
\quad \text{(slow)}
$$

### B. Derivation
For the fast equilibrium:
$$
K_\text{eq} = \frac{[I]}{[A]} = \frac{k_1}{k_{-1}}
\Rightarrow [I] = K_\text{eq}[A]
$$

Slow step controls the overall rate:
$$
r = k_2[I] = k_2 K_\text{eq}[A]
$$

Thus,
$$
r = k_\text{eff}[A], \quad k_\text{eff} = k_2 K_\text{eq}
$$

**Validity condition:**
$$
k_1, k_{-1} \gg k_2
$$

## 1.3. Steady-State Approximation (SSA)

### A.Idea
Intermediate concentration remains **small and nearly constant**:

$$
\frac{d[I]}{dt} \approx 0
$$

### B. Example
$$
A \xrightarrow{k_1} I \xrightarrow{k_2} P
$$

Rate equations:
$$
\frac{d[I]}{dt} = k_1[A] - k_2[I] = 0
\Rightarrow [I] = \frac{k_1}{k_2}[A]
$$

Overall rate:
$$
r = k_2[I] = k_1[A]
$$

### C. More general mechanism
$$
A \xrightarrow{k_1} I, \quad
I \xrightarrow{k_2} P, \quad
I \xrightarrow{k_3} Q
$$

Steady-state for \(I\):
$$
0 = k_1[A] - (k_2 + k_3)[I]
\Rightarrow [I] = \frac{k_1[A]}{k_2 + k_3}
$$

Overall rate of \(P\) formation:
$$
r_P = k_2[I] = \frac{k_1 k_2}{k_2 + k_3}[A]
$$


# 2. Examples

## Example 1

For the reaction of acetaldehyde decomposition the following mechanism is proposed:

$CH_3 CHO \xrightarrow{k_1} CH_3 + CHO$

$CH_3 + CH_3CHO \xrightarrow{k_2} CH_4 + CH_2CHO$

$CH_2CHO \xrightarrow{k_3} CO + CH_3$

$CH_3 + CH_3 \xrightarrow{k_4} C_2H_6$

Using the steady-state approximation, derive the rate equations of formation of $CH_3CHO$ and methane ($CH_4$). 
Consider $CH_3$ and $CH_2CHO$ being unstable intermediates.

### Solution

#### Simplification
Define concentrations: \($[A]=[\mathrm{CH_3CHO}],\; [R]=[\mathrm{CH_3}],\; [X]=[\mathrm{CH_2CHO]}$\).

#### Steady-state approximation
Use the steady-state approximation (SSA) for intermediates \(R\) and \(X\)

Write the rate balances:

$$
\frac{d[R]}{dt} = k_1[A] + k_3[X] - k_2[R][A] - 2k_4[R]^2 \overset{\text{SSA}}{=} 0,
$$

$$
\frac{d[X]}{dt} = k_2[R][A] - k_3[X] \overset{\text{SSA}}{=} 0.
$$

From the \(X\)-balance:

$$
[X] = \frac{k_2}{k_3}\,[R][A].
$$

Substitute into the \(R\)-balance:

$$
0 = k_1[A] + k_3\!\left(\frac{k_2}{k_3}[R][A]\right) - k_2[R][A] - 2k_4[R]^2
= k_1[A] - 2k_4[R]^2.
$$

Hence

$$
[R] = \sqrt{\frac{k_1[A]}{2k_4}}.
$$

Using this in the expression for \([X]\):

$$
[X] = \frac{k_2}{k_3}[A]\sqrt{\frac{k_1[A]}{2k_4}}.
$$

#### Rate of formation of \($\mathrm{CH_3CHO}$\)

Decomposition consumes \(A\) by initiation (step 1) and propagation (step 2):

$$
\frac{d[A]}{dt} = -k_1[A] - k_2[R][A].
$$

Substitute \([R]\):

$$
\frac{d[A]}{dt} = -k_1[A] - k_2[A]\sqrt{\frac{k_1[A]}{2k_4}}
= -k_1[A] - k_2\sqrt{\frac{k_1}{2k_4}}\,[A]^{3/2}.
$$

So the formation rate is

$$
\boxed{\; \dfrac{d[\mathrm{CH_3CHO}]}{dt} = -k_1[\mathrm{CH_3CHO}] \;-\; k_2\sqrt{\dfrac{k_1}{2k_4}}\,[\mathrm{CH_3CHO}]^{3/2}\;}
$$


#### Rate of formation of methane \($\mathrm{CH_4}$\)

Methane is produced only in step (2):

$$
\frac{d[\mathrm{CH_4}]}{dt} = k_2[R][A]
= k_2[A]\sqrt{\frac{k_1[A]}{2k_4}}
= k_2\sqrt{\frac{k_1}{2k_4}}\,[A]^{3/2}.
$$

So

$$
\boxed{\; \dfrac{d[\mathrm{CH_4}]}{dt} = k_2\sqrt{\dfrac{k_1}{2k_4}}\,[\mathrm{CH_3CHO}]^{3/2}\;}
$$

## Example 2

For the mechanism:

\begin{aligned}
&(1)\quad \mathrm{MeBr} \xrightarrow{k_1} \mathrm{Me} + \mathrm{Br}\\
&(2)\quad \mathrm{Me} + \mathrm{MeBr} \xrightarrow{k_2} \mathrm{C_2H_6} + \mathrm{Br}\\
&(3)\quad \mathrm{Br} + \mathrm{MeBr} \xrightarrow{k_3} \mathrm{Me} + \mathrm{Br_2}\\
&(4)\quad 2\mathrm{Me} \xrightarrow{k_4} \mathrm{C_2H_6}
\end{aligned}

Derive the rate equation for formation of ethane ($C_2 H_6$) and molecular bromine ($Br_2$) using the steady-state approximation. Consider Br and Me species being unstable intermediates.

### Solution

#### Short-hand notation:

$[A]=[MeBr],[R]=[Me],[B]=[Br]$

#### Apply the SSA:
Assume \([R]\) and \([B]\) are short-lived intermediates and apply the steady-state approximation (SSA).

$$
\frac{d[R]}{dt} = k_1[A] + k_3[B][A] - k_2[R][A] - 2k_4[R]^2 \overset{\mathrm{SSA}}{=} 0,
$$

$$
\frac{d[B]}{dt} = k_1[A] + k_2[R][A] - k_3[B][A] \overset{\mathrm{SSA}}{=} 0.
$$

From the \(B\)-balance:
$$
k_3[B][A] = k_1[A] + k_2[R][A]
\qquad\Longrightarrow\qquad
k_3[B] = k_1 + k_2[R],
$$
so
$$
\boxed{[B] = \frac{k_1 + k_2 [R]}{k_3}.}
$$

Substitute \($k_3[B][A] = (k_1 + k_2[R])A$\) into the \(R\)-balance:
$$
0 = k_1[A] + (k_1 + k_2[R])A - k_2[R]A - 2k_4[R]^2
= 2k_1[A] - 2k_4[R]^2.
$$

Thus
$$
[R]^2 = \frac{k_1[A]}{k_4}
\qquad\Longrightarrow\qquad
\boxed{[R] = \sqrt{\frac{k_1[A]}{k_4}}\,.}
$$

Using this in the expression for \([B]\) gives
$$
\boxed{[B] = \frac{k_1 + k_2\sqrt{\dfrac{k_1[A]}{k_4}}}{k_3}.}
$$

#### Rates of product formation

**Ethane** \($\mathrm{C_2H_6}$\) is produced by steps (2) and (4):
$$
v_{\mathrm{C_2H_6}} = k_2[R][A] + k_4[R]^2.
$$
Substitute \($[R]^2 = \dfrac{k_1[A]}{k_4}$\) and \($[R]=\sqrt{\dfrac{k_1[A]}{k_4}}$\):
$$
k_4[R]^2 = k_4\cdot\frac{k_1[A]}{k_4} = k_1[A],
$$
so
$$
v_{\mathrm{C_2H_6}}
= k_2[A]\sqrt{\frac{k_1[A]}{k_4}} + k_1[A]
= k_1[A] + k_2\sqrt{\frac{k_1}{k_4}}\,[A]^{3/2}.
$$

**Bromine** \($\mathrm{Br_2}$\) is produced in step (3):
$$
v_{\mathrm{Br_2}} = k_3[B][A].
$$
But from the \(B\)-SSA we have \($k_3[B][A] = k_1[A] + k_2[R][A]$\), hence
$$
v_{\mathrm{Br_2}} = k_1[A] + k_2[A][R]
= k_1[A] + k_2\sqrt{\frac{k_1}{k_4}}\,[A]^{3/2}.
$$

Thus the two product formation rates are identical (consistent with the
overall stoichiometry \($2\mathrm{MeBr}\to\mathrm{C_2H_6}+\mathrm{Br_2}$\):

$$
\boxed{v_{\mathrm{C_2H_6}} = v_{\mathrm{Br_2}} =
\;k_1[A] + k_2\sqrt{\dfrac{k_1}{k_4}}\, [A]^{3/2}.}
$$


#### Remarks
- The term \($k_1[A]$\) originates from initiation (step 1) and termination (step 4) balance; the \($[A]^{3/2}$\) term reflects chain propagation kinetics.
- If additional termination or chain-transfer pathways exist, the coefficient inside the square root will change (replace \($k_4$\) by the effective termination rate).
