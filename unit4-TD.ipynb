#  Unit 4: **1-st law of thermodynamics and thermochemistry: internal energy and enthalpies of phase changes and chemical reactions. Hess’s law.**

# Significance

* **Energy conservation in reactions**
– The First Law ($\Delta U = q + w$) provides the framework for calculating heat exchange and work in chemical systems.

* **Enthalpies of reaction**
– Essential for calorimetry experiments, combustion reactions, and thermochemical tables used in predicting reaction feasibility.

* **Phase change energetics**
– Understanding enthalpies of fusion, vaporization, and sublimation is critical in material synthesis, purification, and cryogenics.

* **Hess’s law**
– Allows calculation of enthalpy changes for reactions that are difficult to measure directly, e.g., lattice enthalpies, enthalpies of formation.

* **Industrial processes**
– Thermochemistry underlies energy balances in large-scale production (ammonia synthesis, combustion furnaces, batteries).

* **Reaction coupling**
– Linking exothermic and endothermic processes to drive unfavorable reactions (e.g., ATP hydrolysis analogy in bioenergetics).

* **Metabolism and bioenergetics**
– The First Law applies to living systems: energy intake (food) → chemical energy (ATP) → work/heat.

* **Phase transitions in biology**
– Enthalpies of melting/freezing explain processes like lipid bilayer transitions, protein denaturation, and cryopreservation.

* **Thermochemistry of drug metabolism**
– Heat release/absorption during biochemical transformations affects stability and pharmacokinetics.

* **Calorimetry in nutrition**
– Hess’s law and combustion enthalpies are used to determine caloric values of foods.

* **Enthalpy in physiology**
– Evaporation enthalpy of water explains cooling via sweating; heat balances are vital in fever, hypothermia, and anesthesia.

* **Medical devices**
– Understanding of thermochemical principles underpins calorimeters, bomb calorimetry in food science, and even aspects of MRI (energy deposition).

## Problem 1

Combustion of 1 g of naphthalene ($C_{10} H_8 (s)$) at 25℃ in a calorimetric bomb yields 40.286 kJ of heat. Water formed in this process is assumed to condense to its liquid state. Compute the standard enthalpy of formation of naphthalene if the standard enthalpies of formation of $CO_2$ and $H_2 O$ are $Δ_f H_{298}^θ ( CO_2 (g))=-393.52 kJ/mol$ and $Δ_f H_{298}^θ (H_2 O(l))=-285.85 kJ/mol$. Report your result in kJ units, rounded up to 2 decimal points.

### Solution

```python
mass = 1.0 # g
M = 10 * 12.0 + 8 * 1.0 # molar mass of naphthalene
n = mass/M  # number of moles

qv = 40286.0 # J energy released to the environment

# Since the process occurs in the bomb calorimeter, the heat released is
# related to the change of internal energy, not enthalpy
dU = -qv # energy change of the system

# Molar internal energy of reaction is 
dUm_r = dU/n
print(F"Molar dU of reaction = {dUm_r} J")

# Write the balanced reaction of combustion:
# C10H8 (s) + (10+2) O2 (g) = 10 CO2 (g) + 4 H2O (l)
# So, the molar enthalpy of reaction is:
# dHm_r = dUm_r + dPV 
# dPV = RTdn, where dn is the change of the moles of gaseous components

dn = 10 - 12 
T = 25 + 273.15 # K
R = 8.314  # J/mol*K
dPV = dn * R * T # J
print(F"Molar dPV of reaction = {dPV} J")

dHm_r = dUm_r + dPV
print(F"Molar dH of reaction = {dHm_r}")

# At the same time, the enthalpy of reaction can be related to enthalpies of formaiton
dHm_f_h2o = -285840.0 # J/mol
dHm_f_co2 = -393520.0 # J/mol
dHm_f_o2 = 0.0

# dHm_r = 10 *  dHm_f_co2 + 4*dHm_f_h2o - dHm_f_x
# So:
dHm_f_x = 10 *  dHm_f_co2 + 4*dHm_f_h2o - dHm_r
print(F"Standard enthalpy of formation of naphthalene = {dHm_f_x} J")
```

## Problem 2

Compute $Δ_r H_{349}^θ$ for the reaction $C_2 H_6 (g)=C_2 H_4 (g)+H_2 (g)$ if the properties are given by the table below. Report your answer in kJ units rounded up to the whole numbers. 

| Substance | $Δ_f H_{298}^θ,kJ mol^{-1}$ |	$C_{p,298},J mol^{-1} K^{-1}$ |	$C_{p,400}, J mol^{-1} K^{-1}$ |
| :-----------: | :---------------------------: | :-----------------------------: | :-------------------------------: |
| C_2 H_6 (g) |	-84.67                      | 52.49                         | 65.44                          |
| C_2 H_4 (g) |	 52.30 | 43.55 | 53.97 |
| H_2 (g) |  N/A   | 28.84 | 29.18 |

### Solution

```python
# Input the parameters:
dH_f_c2h6 = -84670.0 # J/mol*K
dH_f_c2h4 = 52300.0 # J/mol*K
dH_f_h2 = 0.0 # J/mol*K

T1 = 298.0 # K
Cp_298_c2h6 = 52.49 # J/mol*K
Cp_298_c2h4 = 43.55 # J/mol*K
Cp_298_h2 = 28.84 # J/mol*K

T2 = 400.0 # K
Cp_400_c2h6 = 65.44 # J/mol*K
Cp_400_c2h4 = 53.97 # J/mol*K
Cp_400_h2 = 29.18 # J/mol*K

# First, we don't know the heat capacity at a function of T
# we can do it by interpolation:
# C(T) = a + bT
# b = C(T2) - C(T1)/(T2-T1)
# a = C(T1) - b*T1
dT = T2 - T1

b_c2h6 = (Cp_400_c2h6 - Cp_298_c2h6)/dT
b_c2h4 = (Cp_400_c2h4 - Cp_298_c2h4)/dT
b_h2 = (Cp_400_h2 - Cp_298_h2)/dT

a_c2h6 = Cp_298_c2h6 - b_c2h6*T1
a_c2h4 = Cp_298_c2h4 - b_c2h4*T1
a_h2 = Cp_298_h2 - b_h2*T1

# Now, let's compute the heat capacities at the desired temperature of T = 349
"""
T3 = 349.0 # K
Cp_349_c2h6 = a_c2h6 + b_c2h6 * T3
Cp_349_c2h4 = a_c2h4 + b_c2h4 * T3
Cp_349_h2 = a_h2 + b_h2 * T3
"""

# or we could use the average, considering that 349 = 0.5*(298 + 400)
T3 = 349.0 # K
Cp_349_c2h6 = 0.5*(Cp_298_c2h6 + Cp_400_c2h6)
Cp_349_c2h4 = 0.5*(Cp_298_c2h4 + Cp_400_c2h4)
Cp_349_h2 = 0.5*(Cp_298_h2 + Cp_400_h2)

# Compute the dH of reaction at 298:
dH_r_298 = dH_f_c2h4 + dH_f_h2 - dH_f_c2h6

print(F"dH of reaction at 298 K = {dH_r_298} J")

# Now account for the heat capacities and their temperature dependences
dCp = Cp_349_c2h4 + Cp_349_h2 - Cp_349_c2h6
dH_r_349 = dH_r_298 + dCp *(T3 - T1)
print(F"dH of reaction at 349 K = {dH_r_349} J")
```

```python
I_c2h6 = a_c2h6 * (349 - 298) + 0.5*b_c2h6*(349**2 - 298**2)
I_c2h4 = a_c2h4 * (349 - 298) + 0.5*b_c2h4*(349**2 - 298**2)
I_h2 = a_h2 * (349 - 298) + 0.5*b_h2*(349**2 - 298**2)

# Now account for the heat capacities and their temperature dependences
dI = I_c2h4 + I_h2 - I_c2h6
dH_r_349 = dH_r_298 + dI
print(F"dH of reaction at 349 K = {dH_r_349} J")
```
