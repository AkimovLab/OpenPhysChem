# 1. Theory

## 1.1 Key Definitions

### A. Introductory definitions

| **Term** | **Definition** |
|-----------|----------------|
| **Chemical kinetics** | a branch of physical chemistry that studies rates of chemical reactions |
| **Direct kinetic problem** | know reaction mechanisms $\to$ compute rates and kinetic curves (dependence of concentrations on time) |
| **Inverse kinetic problem** | know kinetic curves $\to$ determine mechanisms of reactions  |

The **main object** is: 

$$ aA\ + bB\ + ... \to cC\ + dD\ + ... $$

### B. Comparison to thermodynamics

|           |  **Kinetics**    |  **Thermodynamics** |
| ---       | ---              | ---                 | 
| **Main question** | How fast?    | What's in the end?  |
| **Time**          | Is the main variable | No time  |
| **Path**          | Reaction doesn't proceed as written (only stoichiometry). In reality, there are intermediates. Main question: mechanisms of chemical reactions (what happens in the “black box”?) | Path is "not important" |


### C. Reaction Mechanisms and Elementary Reactions

**Definition: Mechanism of a chemical reaction** is a series of elementary reactions that can be physically realized in the reactive mixture of reagents and products under given conditions.

- Mechanism may involve not only reactants, but also products

- Mechanism may be different under different conditions

**Reactions classify as:**
- Simple (elementary) - involve only one step
- Complex - involve more than one step

**Elementary reaction** can be defined in several ways:

1. Definition 1 (simple):  The reaction in which only **one bond is broken or formed**

2. Definition 2 (Tolmen): The reaction which has **one potential energy barrier** 

3. Definition 3 (IUPAC): In which one or more chemical species react directly to form products in **a single reaction step** and with **a single transition state**.


**Molecularity** - minimal number of particle necessary to accomplish reaction (can nt be larger than 3)

- $A \to ...$ - monomolecular 
- $A + B \to ...$ - bimolecular (A and B can be the same kinds of molecules)
- $A + B + C \to ...$ - trimolecular (A, B, and C - all can be the same kinds of molecules) 

**Notation:**

- $=$ is used to denote stoichiometry
- $\to$ is used to denote elementary reactions (mechanisms)


## 1.2  Reaction Rates

### A. Definitions 

**Rate of chemical reaction $r_j$** – the number of elementary acts happening per unit time per unit volume of reaction space.
For instance, for j-th reaction (assuming there are many channels) 
$$
aA + bB + ... = cC + dD
$$
the rates are:

$$
r_j = \frac{1}{a} W_A = \frac{1}{b} W_B = \frac{1}{c} W_C = \frac{1}{d} W_D
$$

Here, $W_X$ is the speed of reactant consumption or product formation, defined as:

$$
W_A = - \frac{d [A]}{dt} \\ 
W_B = - \frac{d [B]}{dt} \\ 
W_C = + \frac{d [C]}{dt} \\ 
W_D = + \frac{d [D]}{dt} \\ 
$$

Here, $[X]$ is the concentration of component X

**Notice the sign:** the idea is that $r_j$ should be positive (as the number of acts of reaction is always non-negative),
but the change of concentration ($\frac{d [X]}{dt}$) is different for reactants (decreasing, so the $W_X$ is negative), and
products (increasing, so the $W_X$ is positive)

### B. Rate of product/reactant change

The speed of reactant disappearance/product formation of a component $A_i$ is given by summing the contributions from all reactions (also called reaction channels, indexed by integer $j$) that take place in the system in parallel:

$$
\boxed{ \frac{d[A_i]}{dt} = \sum_j { a_{ij} r_{j} } + D_{A_i} \frac{\partial^2 [A_i]}{\partial x^2}
}
$$
Here,
- $a_{ij}$ is a stoichiometric coefficient of compoent $A_i$ in reaction j: $a_{ij} < 0$ if $A_i$ is a reactant and $a_{ij} > 0$ is a product in this reaction; 
- the first term on the RHS describes changes in concentrations due to reactions;
- the second term on the RHS describes the diffusion of component $A_i$; The inclusion of gradients is especially important in crowded environments (e.g. reactions in proteins, etc.), when viscosity is high, and in large-scale systems (e.g. in industrial setups – this may be essential!).

in typical PChem course, the diffusion term is usually neglected, so we use:

$$
\boxed{ \frac{d[A_i]}{dt} = \sum_j { a_{ij} r_{j} } 
}
$$

**Stoichiometric basis** - the set of linearly-independent elementary reactions sufficient to describe the overall reaction of interest and that accounts for all key intermediates. 

**Example (Bodenstein):**

For reaction 
$$
H_2 + Br_2 = 2 HBr
$$

one may consider the following elementary reactions:
$$
Br_2 \to 2Br\quad (1) \\
Br + H_2 \to HBr + H\quad (2) \\
H + Br_2 \to HBr + Br\quad (3) \\
Br_2 + H_2 \to 2HBr\quad (4) \\
Br + H \to HBr\quad (5) \\
HBr + H \to H_2 + Br\quad (6) \\
H_2 \to 2H\quad (7) \\
2H \to H_2\quad (8) \\
2Br \to Br_2\quad (9) \\
HBr + Br \to H + Br_2\quad (10) 
$$

However, some equations are linearly-dependent, e.g. Eq. 9 is the reverse of Eq. 1, Eq. 8 is the reverse of Eq. 7, Eq. 2 +  Eq. 3 = Eq. 4, etc. The minimal basis could be chosen as:

$$
Br_2 \rightleftharpoons 2Br\quad (1) \\
Br + H_2 \rightleftharpoons HBr + H\quad (2) \\
H + Br_2 \rightleftharpoons HBr + Br\quad (3)
$$

### C. Kinetic mass action law, rate orders

The **kinetic mass action rate law** defines the way to compute rates of reactions:

$$
r_j = k_j [A]^{n_A} [B]^{n_B} ...
$$

Here:
- $k_j$ is the **rate constant** in reaction $j$
- $n_A, n_B$ are **reaction orders** with respect to componets A, B, etc., respectively
- $n = n_A + n_B + ... $ - the **total reaction order** 

**Kinetic rate equation** – equation that connects rate of chemical reaction and concentrations of substances that participate in the reaction. (Participate is understood in a broad sense: e.g. catalysts). 

There are two types of kinetic rate equations:

- **Model kinetic rate equation:**
$$
aA + bB + ... \to zZ + ...\quad r_j = k_j [A]^a [B]^b ...
$$
1. $a, b, ...$ - are integers, called **model reaction orders**
2. $a + b + ...$ - reaction molecularity (usually not larger than 3)
3. only reactants’ concentrations are involved in the kinetic rate equation

- **Experimental kinetic rate equation:**
$$
aA + bB + ... \to zZ + ...\quad r_j = k'_j [A]^{n_a} [B]^{n_b} [Z]^{n_z} \times ...
$$
1. $a, b, ...$ - can be fractional numbers, called **apparent reaction orders**
2. Apparent orders can be used as a diagnostic tool: if integer and if there is no dependence on products then the reaction is elementary
3. both reactants' and products' concentrations may be involved in the kinetic rate equation



## 3. Kinetics of Simple Reactions

**Half-life $(\tau_{1/2})$** - time needed to complete the reaction by 50% (half). In other words, it is the time when the current concentration of the component is half of its initial amount.

### A. Zero-Order Reaction

$$ A \xrightarrow{k} P $$

**Rate law:**  
$$ -\frac{d[A]}{dt} = r = k $$

**Integrated form:**  
$$ [A] = [A]_0 - kt $$

**Half-life:**  
$$ \tau_{1/2} = \frac{[A]_0}{2k} $$

Rate is independent of concentration.


### B. First-Order Reaction

$$ A \xrightarrow{k} P $$

**Rate law:**  
$$ -\frac{d[A]}{dt} = r = k[A] $$

**Integrated form:**  
$$ \ln\frac{[A]_0}{[A]} = kt $$

or 

$$ [A](t) = [A]_0 \exp(-kt) $$

**Half-life:**  
$$ \tau_{1/2} = \frac{\ln 2}{k} $$

**Examples:** radioactive decay, unimolecular isomerization.


### C. Second-Order Reaction

#### Case 1: 

$$ 2A \xrightarrow{k} P $$

**Rate law:**  
$$ -\frac{d[A]}{dt} = r = k[A]^2 $$

**Integrated form:**  
$$ \frac{1}{[A](t)} = \frac{1}{[A]_0} + kt $$

**Half-life:**  
$$ \tau_{1/2} = \frac{1}{k[A]_0} $$


#### Case 2:
$$ A + B \xrightarrow{k} P $$

**Rate law:**  
$$ -\frac{d[A]}{dt} = r = k[A] [B]$$

**Integrated form:**  
if $[A]_0 = [B]_0$, then:
$$ \frac{1}{[A](t)} = \frac{1}{[A]_0} + kt $$
$$ \tau_{1/2} = \frac{1}{k[A]_0} $$

if $[A]_0 \neq [B]_0$, then:
$$ kt = \frac{1}{[A]_0 - [B]_0} \ln { \frac{[A](t)[B]_0}{[B](t)[A]_0} } $$
$$ \tau_{1/2}(A) \neq \tau_{1/2}(B)$$


### D. n-th-Order Reaction

$$ nA \xrightarrow{k} P\quad n > 1 $$

**Rate law:**  
$$ -\frac{d[A]}{dt} = r = k[A]^n $$

**Integrated form:**  
$$ kt = \frac{1}{n-1} \left( \frac{1}{[A]^{n-1}(t)}  - \frac{1}{[A]^{n-1}_0} \right )  $$

**Half-life:**  
$$ \tau_{1/2} = \frac{2^{n-1}-1}{k(n-1)[A]^{n-1}_0} $$


### E. Pseudo-First-Order Reaction

If one reactant is in large excess (constant concentration),  
a higher-order reaction becomes **effectively first-order**.

**Example:**  
$$ \text{CH}_3\text{COOC}_2\text{H}_5 + \text{H}_2\text{O} \rightarrow \text{CH}_3\text{COOH} + \text{C}_2\text{H}_5\text{OH} $$

Water is in large excess ⇒ pseudo-first order in ester.

## Example 1 [Eremin Example 1-1]

The rate of formation of NO in reaction $2NOBr(g) \to 2NO(g) + Br_2(g)$ is $1.6\cdot 10^{-4} M\cdot s^{-1}$. What is the rate of reaction and what is the rate of consumption of NOBr?

### Solution
By definition:
$$
w = -\frac{1}{2} \frac{d[NOBr]}{dt} = \frac{1}{2} \frac{d[NO]}{dt} = \frac{1}{2} $1.6\cdot 10^{-4} M\cdot s^{-1} = 
$8.0\cdot 10^{-5} M\cdot s^{-1}
$$

From here, the rate of NOBr consumption is the negative of the rate of the NO formation:
$$
\frac{d[NOBr]}{dt} = - \frac{d[NO]}{dt} = -1.6\cdot 10^{-4} M\cdot s^{-1}
$$

## Example 2 [Eremin Example 1-2]
In a second-order reaction $A + B \to D$, the initial concentrations of A and B are 2 M and 3 M, respectively. The rate of reaction is $1.2\cdot 10^{-3} M/s$ when concentration of A is $[A] = 1.5 M$. Compute: a) the reaction rate constant; b) rate of reaction when $[B] = 1.5 M$

### Solution
When $[A](t) = 1.5 M$, it means that $[A]_0 - [A](t) = 2 - 1.5 = 0.5 M$ has reacted. According to the stoichiometry, this means that 0.5 M of B has reacted as well, so its current concentration is $[B](t) = [B]_0 - 0.5 = 3 - 0.5 = 2.5$ 

From $r(t) = k [A](t) [B](t)$, we obtain $k = \frac{r(t)}{[A](t)[B](t)} = \frac{ 1.2\cdot 10^{-3} M/s}{1.5 M \cdot 2.5 M}= 3.2\cdot 10^{-4} M^{-1} s^{-1}$

When $[B](t) = 1.5$ (3 - 1.5 = 1.5 M of B reacted), the concentration of A is: $[A](t) = 2 - 1.5 = 0.5 M$, so the rate is:
$$
r = k [A] [B] = 3.2\cdot 10^{-4} M^{-1} s^{-1}\cdot 0.5 M\cdot 1.5 M = 2.4 \cdot 10^{-4} M s^{-1}
$$


## Example 3 [Eremin Example 2-1]

Half-life of radioactive isotope of carbon $^{14}C$ is 5730 years. During an archeological excavation a tree was found that contained 72% compared to normal content. What is the tree's age?

### Solution
For radioactive decay, we use the first-order kinetics, so:
$$
k = \frac{\ln 2}{\tau_{1/2}}
$$
and
$$
[A](t) = [A]_0 \exp(-kt)
$$

```python
import math
#
tau = 5730 # years
k = math.log(2.0)/tau # year^{-1}

# A = 0.72 A_0, so
# 0.72 = exp(-k*t) => t = -log(0.72)/k
t = -math.log(0.72)/k

print(F"The age is {t:.0f} years")
```

## Example 4

The second-order reaction $A + B \to P$ is carried out in a solution with initial concentrations $[A]_0 = 0.050\ M$ and $[B]_0 = 0.080\ M$. After 1 hour, the concentration of substance A decreased to 0.020 M. Calculate the rate constant and half-lives of both substances.

### Solution

We use the formula for $[A]_0 \neq [B]_0$:

$$
k\ t = \frac{1}{[A]_0 - [B]_0} \ln \left ( \frac{[A][B]_0}{[B][A]_0} \right )
$$

```python
import math
import matplotlib.pyplot as plt
import numpy as np

A0 = 0.05 # M
A = 0.02 # M

# How much of x reacted:
x = A0 - A

B0 = 0.080 # M
B = B0 - x  # the amount of consumed B is as much as A

# 
t = 1.0 # h
k = math.log(A * B0 / (A0*B)) /(t * (A0 - B0) )

print(F"k = {k} mol/(L*h) = {k/60.0} mol/(L*min)")

# To compute half-lives, we set the values of the current concentrations of A or B to half of the initial values
# First for A
A = 0.5 * A0
x = A0 - A
B = B0 - x

# Using the same equation as before, but solve for t:
tA = math.log(A * B0 / (A0*B)) /(k * (A0 - B0) )
print(F"Half-life of A is = {tA} = {tA * 60.0} min")

# Next for B
B = 0.5 * B0
x = B0 - B
A = A0 - x

# Using the same equation as before, but solve for t:
tB = math.log(A * B0 / (A0*B)) /(k * (A0 - B0) )
print(F"Half-life of B is = {tB} h = {tB * 60.0} min")
```
