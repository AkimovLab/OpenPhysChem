# 1. Theory

Convention on notation:
- A – solvent (major component)
- B – solute
- J – generic component 

## 1. Raoult’s Law

> Partial pressure of a substance J in a liquid mixture is proportional to mole fraction of this substance and the vapor pressure of the substance when pure. $$\boxed{p_J = x_J p_J^{\ast} }$$

**Significance:**
- Applies to **solvent** (major component).  
- Deviations indicate non-ideal behavior due to solute–solvent interactions.  
- Consequence: $\mu_A(P) = \mu_A^{\theta} + RT\ln(x_A)$

**Ideal solution** – when Raoult’s law is satisfied for all compositions of the mixture/solution

For an **ideal binary liquid solution**:

$$
p_A = x_A p_A^{\ast}, \quad p_B = x_B p_B^{\ast}
$$

- $p_A, p_B$: partial vapor pressures  
- $x_A, x_B$: mole fractions in liquid phase  
- $p_A^{\ast}, p_B^{\ast}$: vapor pressures of pure components  

- **Works best when:** Components have similar molecular shapes and are held together by similar types of forces
- **Deviations are small:** for the component that is in large excess (solvent)
- **Applicable to:** vapor pressure of solvent in a dilute solution


## 2. Henry’s Law

> Partial pressure of a **volatile solute, B,** is proportional to mole fraction of this substance in a solution. The proportionality factor is different from pressure of the pure substance, the **Henry’s constant**. 

**Significance:**
- Relates gas solubility in a liquid to the gas partial pressure.
- Applies to **solute** (minor component, typically a gas).  
- Consequence - chemical potential of solute: $\mu_B([B]) = \mu_B^{\theta} + RT\ln([B])$

**Mathematically:**

$$P_B = x_B K_H$$

or

$$P_B = [B] K'_H$$

Where:
- $[B]$: concentration of solute (e.g. dissolved gas)
- $P_B$: partial pressure of solute above solution  
- $k_H$: Henry’s constant (depends on solute, solvent, and temperature)  


## 3. Real Solutions and Activities

For **non-ideal solutions**, we introduce the **activity**, $a_J$, which replace pressures for gases and concentrations for liquid or dissolved components:

$$
p_J \rightarrow a_J \quad
[J] \rightarrow a_J
$$

so, generally:

$$
\mu_J = \mu_J^{\theta} + RT\ln(a_J)
$$

where

$a_A = \gamma_A x_A$ - for solvent
$a_B = \gamma_B [B]$ - for solute

Here,
- $a_J$: activity of component J
- $\gamma_J$: activity coefficient of component J (measures deviation from ideality)  
- $x_J$: mole fraction  
- $[J]$: concentration

**Limiting properties of Activity Coefficients:**
- $\gamma = 1$ → ideal solution.  
- $\gamma > 1$ → positive deviation (weaker solute–solvent interactions).  
- $\gamma < 1$ → negative deviation (stronger solute–solvent interactions).  
- $\gamma_A \rightarrow 1 \text{as} x_A \rightarrow 1$ - dilute solution 
- $\gamma_B \rightarrow 1 \text{as} [B] \rightarrow 0$ - dilute solution
- $\gamma_J = 1$ - for pure liquids and solids


## 4. Colligative Properties

Colligative properties depend only on the **number of solute particles**, not their identity.  
For electrolytes, include the **van ’t Hoff factor \(i\)** to account for ionization.

1. **Freezing point depression**:
   $$
   \Delta T_f = i K_f \, m_B
   $$

2. **Boiling point elevation**:
   $$
   \Delta T_b = i K_b \, m_B
   $$

   $$
   K_b = \frac{R\ T_b^*}{\Delta_{vap} H_m}
   $$

3. **Osmotic pressure**:
   $$
   \Pi_B = i M_B R T
   $$

- $m_B$: molality of solute ($\frac{\#\ 𝑜𝑓\ 𝑚𝑜𝑙𝑒𝑠\ 𝑜𝑓\ 𝑠𝑜𝑙𝑢𝑡𝑒}{𝑘𝑔\ 𝑜𝑓\ 𝑠𝑜𝑙𝑣𝑒𝑛𝑡}$)
- $M_B$: molarity of solute  
- $K_b, K_f$: ebullioscopic and cryoscopic constants of solvent  
- $i$: van ’t Hoff factor (number of particles formed per solute unit) 
- $T_b^*$: normal boiling temperatue
- $\Delta_{vap} H_m$ - molar enthalpy of vaporization of solvent


5. **Schröder–van Laar Equation**

> Describes **liquidus curves** (melting point depression) in binary phase diagrams of ideal mixtures.  
From equilibrium between solid and liquid chemical potentials:

$$
\ln x_i = - \frac{\Delta H_{f,i}}{R} \left( \frac{1}{T} - \frac{1}{T_{m,i}} \right)
$$

- $x_i$: mole fraction of component $i$ in liquid  
- $T$: **liquidus** temperature of mixture - the temperature above which the mixture is completely liquid at a given composition.
- $T_{m,i}$: melting point of pure $i$  
- $\Delta H_{fus,i}$: molar enthalpy of fusion  
- $R$: ideal gas constant  

**Application**:  
- Predicts melting point depression.  
- Basis for constructing **eutectic phase diagrams**.

## Problem 1

Ethanol and methanol form an almost ideal solution when mixed together. At 20 ℃, the vapor pressure of pure ethanol is 5.93 kPa and pure methanol is 11.83 kPa. Compute the total pressure of the vapor that is in equilibrium with a mixture of 100 g of ethanol and 100 g of methanol. Also, compute the composition (molar fractions of components) of such a vapor. 

### Solution

```python
# The conditions given

P1 = 5.93 # kPa
P2 = 11.83 # kPa
m1 = 100.0  # g of ehtanol C2H5OH
m2 = 100.0  # g of methanol CH3OH

# Molar masses
M1 = 2 * 12 + 6 + 16   # C2H5OH
M2 = 12 + 4 + 16       # CH3OH

# Number of moles of components in the mixture
n1 = m1/M1  # ethanol
n2 = m2/M2  # methanol
n = n1 + n2 # total

# Molar fractions of components in liquid phase
x1 = n1/n
x2 = n2/n

# The total pressure - Raoult's law
P = x1 * P1 + x2 * P2

# Composition of vapor - via partial pressures
X_eth = P1 * x1 / P

print(F"Total pressure = {P} kPa, fraction of ehtanol = {X_eth}, fraction of methanol = {1-X_eth }" )
```

## Problem 2

A solution containing 0.217 g of sulphur dissolved in 19.18 g of CS2 boils at 319.304 K. The boiling temperature of pure CS2 is 319.2 K. The ebulioscopic constant of CS2 is 2.37 $\frac{K*kg}{mol}$. How many sulphur atoms an in the sulphur molecule dissolved in CS2?

### Solution

We just apply the formula:
$\Delta T_b = K_b b_B$ 

```python
K_b = 2.37 # K*kg/mol
dT_b = 319.304 - 319.2 # raising of the boling point
m_solvent = 19.18e-3 # kg of solvent
m_S = 0.217 # g of solute

# Compute the molality:
b = dT_b/K_b

# Molality is the number of moles of solute per kg of solvent
n_S = b * m_solvent  # the number of moles of sulphur molecule

# n = m/M, so the molar mass of S molecule is:
M = m_S / n_S

print(F"Molar mass of S_n molecule = {M} g/mol")

# Molar mass of sulphur atom:
M_S = 32.065 # g/mol

# So, the number of S atoms in  S_n molecule is:
n = M/M_S

print(F"ratio = {n}, the number of atoms is {  int(n) }")
```
