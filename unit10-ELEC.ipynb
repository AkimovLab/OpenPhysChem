# 1. Theory

## 1.1 Terminology

- **Electrolytes**: Substances that conduct electricity when dissolved in water, due to the presence of ions.  
- **Electrodes**: Interfaces where oxidation or reduction occurs.  
- **Electrochemical potential (μ̃)**: Sum of chemical potential and electrical potential contribution. Governs ion motion in an electric field.  
- **Conductivity (κ)**: Proportional to ion concentration and mobility.  
- **Electrochemical equilibrium**: Established when the electrochemical potentials of species are balanced across an interface.  
---

## 1.2 Debye–Hückel Theory

Describes how electrostatic interactions affect ion behavior in dilute solutions.  

- **Key idea**: Each ion is surrounded by an “ionic atmosphere” of opposite charge, which reduces its effective activity.  
- **Activity coefficient (γ)**: Corrects concentrations to account for non-ideal ionic interactions.  
- **Debye–Hückel limiting law** (valid at low ionic strength *I*):  

$$
\log \gamma_\pm = -A z^2 \sqrt{I}
$$

where:  
- *γ±* = mean activity coefficient  
- *z* = ion charge  
- *I* = ionic strength  
- *A* = constant depending on temperature and solvent dielectric constant  
- **Consequence**: Ionic solutions deviate from ideality even at low concentrations due to long-range Coulomb interactions.  

### Ionic Strength

The **ionic strength** (*I*) of a solution is a measure of the total concentration of ions, weighted by the square of their charges.  

**Formula:**  

$$
I = \tfrac{1}{2} \sum_{i} c_i z_i^2
$$  

where:  
- \($c_i$\) = molar concentration of ion *i* (mol·L⁻¹)  
- \($z_i$\) = charge number of ion *i*  
- The factor \($\tfrac{1}{2}$\) prevents double counting of ion–ion interactions.  

### Expression for A 
In general:

$$
A = \frac{1}{3} \left( \frac{2 \pi N_A \rho}{1000} \right)^{1/2} \cdot \frac{e^3}{(8 \pi \varepsilon_0 \varepsilon_r k_B T)^{3/2} \cdot 2.303 \, R T}
$$  

where:  
- \($N_A$\) = Avogadro’s number  
- \($\rho$\) = solvent density (kg·L⁻¹)  
- \(e\) = elementary charge  
- \($\varepsilon_0$\) = permittivity of free space  
- \($\varepsilon_r$\) = dielectric constant of solvent  
- \($k_B$\) = Boltzmann constant  
- \(R\) = gas constant  
- \(T\) = absolute temperature  


In practice:
For **aqueous solutions at 25 °C**:

$$
A \approx 0.509 \; \text{(mol/L)}^{-1/2}
$$  

- At 18 °C, \($A \approx 0.488$\)  
- At 100 °C, \($A \approx 0.73$\)

## 1.3 Oxidation and Reduction

- **Redox reactions**: Chemical processes involving electron transfer.  
  - **Oxidation** = loss of electrons  
  - **Reduction** = gain of electrons  
- Always occur in pairs (oxidation + reduction).  

### Oxidation
- **Definition**: The process of **loss of electrons** by a species.  
- General form:  

$$
\text{M} \;\;\rightarrow\;\; \text{M}^{n+} + n e^-
$$

- Example (Zinc in Daniell cell):  

$$
\text{Zn} \;\;\rightarrow\;\; \text{Zn}^{2+} + 2 e^-
$$  


### Reduction
- **Definition**: The process of **gain of electrons** by a species.  
- General form:  

$$
\text{X}^{n+} + n e^- \;\;\rightarrow\;\; \text{X}
$$

- Example (Copper in Daniell cell):  

$$
\text{Cu}^{2+} + 2 e^- \;\;\rightarrow\;\; \text{Cu}
$$  

---


## 1.4 Electrochemical Cells

### Electrochemical Cells

1. **Galvanic (Voltaic) cells**  
   - Spontaneous redox reactions produce electricity.  
   - Example: Daniell cell (Zn | Zn²⁺ || Cu²⁺ | Cu).  

2. **Electrolytic cells**  
   - External electrical energy drives non-spontaneous reactions.  
   - Example: Electrolysis of water.  
   
### Electrodes

- **Definition**: Electrodes are conductive materials (usually metals or carbon) that provide a surface for electron transfer in an electrochemical cell.  
- Two types in a cell:  
  - **Anode** → where **oxidation** occurs  
  - **Cathode** → where **reduction** occurs  

👉 Mnemonic: **“AnOx, RedCat”**  
- **An**ode = **Ox**idation  
- **Red**uction at **Cat**hode  

### Recap 

- **Anode** = site of oxidation (electrons leave the system)  
- **Cathode** = site of reduction (electrons enter the system)  
- Electrons flow **anode → cathode** through the external circuit.  

### Cell Potential and Nernst Equation  

- **Standard electrode potential (E°)**: Measure of a half-reaction’s tendency to be reduced.  
- **Cell potential (Ecell)**:  

$$
E_{\text{cell}} = E^\circ_{\text{cathode}} - E^\circ_{\text{anode}}
$$  

- **Nernst equation** (accounts for non-standard conditions):  

$$
E = E^\circ - \frac{RT}{nF} \ln Q
$$

where:  
- *n* = number of electrons transferred  
- *F* = Faraday constant  
- *Q* = reaction quotient

## Problem 1 (computing ionic strengths)

Compute ionic strength in the following solutions:

a) NaCl solution (0.10 M NaCl)
b) CaCl₂ solution (0.10 M CaCl₂)
c) Mixture of NaCl (0.05 M) and MgSO₄ (0.10 M)

### Solution

a) NaCl solution (0.10 M NaCl)

- NaCl → Na⁺ (0.10 M, charge +1), Cl⁻ (0.10 M, charge –1)  

$$
I = \tfrac{1}{2} \Big[ (0.10)(+1)^2 + (0.10)(-1)^2 \Big]
$$  

$$
= \tfrac{1}{2} (0.10 + 0.10) = 0.10 \;\text{M}
$$  


b) CaCl₂ solution (0.10 M CaCl₂)

- CaCl₂ → Ca²⁺ (0.10 M, charge +2), Cl⁻ (0.20 M, charge –1)  

$$
I = \tfrac{1}{2} \Big[ (0.10)(+2)^2 + (0.20)(-1)^2 \Big]
$$  

$$
= \tfrac{1}{2} (0.40 + 0.20) = 0.30 \;\text{M}
$$  


c) Mixture of NaCl (0.05 M) and MgSO₄ (0.10 M)

- NaCl → Na⁺ (0.05 M, +1), Cl⁻ (0.05 M, –1)  
- MgSO₄ → Mg²⁺ (0.10 M, +2), SO₄²⁻ (0.10 M, –2)  

$$
I = \tfrac{1}{2} \Big[ (0.05)(+1)^2 + (0.05)(-1)^2 + (0.10)(+2)^2 + (0.10)(-2)^2 \Big]
$$  

$$
= \tfrac{1}{2} (0.05 + 0.05 + 0.40 + 0.40) = \tfrac{1}{2} (0.90) = 0.45 \;\text{M}
$$  


## Problem 2

Using the Debye-Huckel law, compute the activity coefficients of ions $K^+$, $Al^{3+}$ and $SO_4^{2-}$ in a $1.0 \times 10^{-3}$  M solution of $KAl(SO_4 )_2$ at 25 ℃.

### Solution

```python
# It is useful to use some numpy:
import numpy as np

# Compute the molality of the solution: 1 L of water = 1 kg, so mol/L is numerically equal to mol/kg, so we can take 
# the molality of solution to be 1.0 * 10^-3 mol/kg
b_sol = 1e-3 # mol/kg

# Molalities of ions:
#                  K           Al        SO4
b = np.array( [ 1*b_sol,    1*b_sol,  2*b_sol])

# Charges of ions:
#              K+           Al3+      SO4^2- 
z = np.array([1,            3,       -2])

# Ionic strength:
I = 0.5 * np.sum( z*z * b)
print(I)

# Activity coefficients
A = 0.5

gamma  = 10 ** (-A*z*z*np.sqrt(I))
print(gamma)
```

## Problem 2

Compute the equilibrium constant of the disproportionation reaction $2Cu^+ ↔ Cu^{2+} + Cu(s)$ at 25 ℃. Use the tabulated date for standard redox potentials (V)



### Solution

$Cu^+ + e^- -> Cu$   $E_1 = 0.521$ (R1)

$Cu^{2+} + 2e^- -> Cu$  $E_2 = 0.337$  (R2)

The $\Delta G$ for these reactions are:

$\Delta G_1 = -1 F E_1$ (one electron process)

$\Delta G_2 = -2 F E_1$ (two electron process)

The total reaction:

$2*(R1) - (R2) = 2Cu^+ + 2e^- + Cu -> Cu^{2+} + 2e^- + 2Cu <=>  2 Cu^+ -> Cu^{2+} + Cu$ 

Hence:

$\Delta G = 2\Delta G_1 - \Delta G_2$

```python
F = 96485 # C/mol
R = 8.314
T = 25 + 273.15

E1 = 0.521 # V
E2 = 0.337 # V

dG1 = -1*F*E1
dG2 = -2*F*E2
dG = 2*dG1 - dG2 

print(F"dG = {dG} J/mol")

# Equilibrium constant:
lnK = -dG / (R * T)
K = np.exp(lnK)

print(F"K = {K}")
```
