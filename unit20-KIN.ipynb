# 1. Theory

## 1.1. TAC: Theory of active collisions

The **Theory of Active Collisions (TAC)** is an extension of collision theory that accounts not only for the number of collisions but also for their **energetic** and **geometric (steric/orientational)** effectiveness.  

It refines the simple collision theory by introducing **active collisions** – only a fraction of all collisions lead to reaction.


### A. General rate constant expression

$$
k_{TAC}(T) = P_{AB}\ Z_{AB} \ e^{-E_a/(RT)}
$$

- \($Z_{AB}$\): collision frequency between species \(A\) and \(B\)  
- \($P_{AB}$\): steric (orientation) factor ($0< P_{AB}<1$)  
- \($E_a$\): true activation energy  
- \(R\): gas constant, \(T\): temperature  

### B. Collision frequency
For a **bimolecular gas-phase reaction**:

$$
Z_{AB} = N_A \, \sigma_{AB} \, \langle v_{\text{rel}} \rangle \, [A][B]
$$

- \($N_A$\): Avogadro’s number  
- \($\sigma_{AB} = \pi (d_A + d_B)^2$\): collision cross-section  
- \($\langle v_{\text{rel}} \rangle$\): mean relative speed  

$$
\langle v_{\text{rel}} \rangle = \sqrt{\frac{8RT}{\pi \mu}}
$$

where \($\mu = \frac{m_A m_B}{m_A+m_B}$\) is the reduced mass.


### C. Steric (orientation) factor

- Accounts for **molecular geometry and orientation requirements**.  
- Typical values range from \($10^{-3}$\) to \($10^{-8}$\) for complex organic reactions.  

$$
P_{AB} = \frac{k_{\text{exp}}}{Z_{AB} e^{-E_a/(RT)}}
$$

### D. Combined Equation: Trautz–Lewis Equation 

The bimolecular gas-phase reaction rate constant is:

$$
\boxed{ k
= P\ N_A\ d_{AB}^2 \left( \frac{8\pi k_B T}{\mu_{AB}} \right)^{1/2} \exp(-E_a/(k_B T))
}
$$

where

- $N_A$ is Avogadro's number  
- $d_{AB} = (d_A + d_B)$ is the minimal distance between atoms 
- $\mu_{AB}$ is the reduced molar mass  
- $P$ is the steric factor  
- $E_a$ is the (true) activation energy  
- $k_B$ is the Boltzmann constant  
- $T$ is the temperature  


### E. Arrhenius form in TAC

The TAC framework naturally leads to the Arrhenius equation:

$$
k(T) = A \, e^{-E_a/(RT)}, \quad A = Z_{AB}\ P_{AB} 
$$

- \(A\) (pre-exponential factor) is explained as **collision frequency × orientation probability**.  
- \($E_a$\) is the true activation barrier.


### F. Key Points

1. **Not all collisions are reactive** → only “active collisions” lead to reaction.  
2. **Two main corrections to simple collision theory**:  
   - Energy requirement \($E \geq E_a$\)  
   - Orientation/steric requirement \($P_{AB}$\)  
3. Provides a **microscopic basis** for the Arrhenius parameters \(A\) and \($E_a$\).  
4. Bridges **collision theory** and **transition-state theory (TST)** by emphasizing both dynamical and structural constraints.

## 1.2. Lindemann–Hinshelwood Mechanism of unimolecular reactions

The mechanism for a unimolecular reaction is:

Activation step:

$$
A + M \overset{k_1}{\underset{k_{-1}}{\rightleftharpoons}} A^* + M
$$

Decay step:

$$
A^* \xrightarrow{k_2} \text{products}
$$

Effective rate constant (steady-state on $A^*$):

$$
k_{\text{eff}}
= \frac{k_1 k_2 [M]}{k_{-1} + k_2 + k_1[M]}.
$$

Low-pressure limit:

$$
k_{\text{eff}}
= \frac{k_1 k_2 [M]}{k_{-1} + k_2}.
$$

Pressure-dependent form:

$$
k_{\text{eff}} = \frac{k_0 [M]}{1 + \dfrac{k_0 [M]}{k_\infty}},
$$

where  
$k_0 = \dfrac{k_1 k_2}{k_{-1} + k_2}$ and  
$k_\infty = k_2$.



## 1.3. TST: Transition state theory (activated complex theory)

### The key results 

The reaction proceeds via a quasi-equilibrium with the transition state:

$$
A + B \rightleftharpoons [AB]^\ddagger
$$

Canonical TST rate constant:

$$
k(T) = \kappa \frac{k_B T}{h} \frac{Q^\ddagger(T)}{Q_A Q_B} e^{-E_0/(k_B T)}
$$

where  
- $\kappa$ = quantum transmission coefficient ($\kappa \approx 1$ for adiabatic and $\kappa << 1$ for nonadiabatic regimes)
- $k_B$ = Boltzmann constant  
- $h$ = Planck constant  
- $Q^\ddagger$, $Q_A$, $Q_B$ = partition functions of the transition state and reactants  
- $E_0$ = activation energy

Assumptions:  
- Classical motion along the reaction coordinate  
- No recrossing of the transition state  
- Reactants in thermal equilibrium

Connection to Arrhenius equation:

$$
k(T) = A \, e^{-E_a / (RT)}, \quad 
A \sim \frac{k_B T}{h} \frac{Q^\ddagger(T)}{Q_A Q_B}
$$

### The derivations of the exponential factor

* **Step 1 — Partition function of reactants**

For a reactant (or reactants $A + B$):

$$
Q_\text{reactants} = \sum_j e^{-E_j^\text{react}/(k_B T)}
$$

Let the **lowest energy of the reactants** be $E_\text{R}$. Factor it out:

$$
Q_\text{reactants} = e^{-E_\text{R}/(k_B T)} \sum_j e^{-(E_j^\text{react} - E_\text{R})/(k_B T)}
$$

* **Step 2 — Partition function of the transition state**

Similarly, for the transition state (TS):

$$
Q^\ddagger = \sum_i e^{-E_i^\ddagger/(k_B T)} 
= e^{-E_\text{TS}/(k_B T)} \sum_i e^{-(E_i^\ddagger - E_\text{TS})/(k_B T)}
$$

where $E_\text{TS}$ is the **minimum potential energy of the TS**.

* **Step 3 — Ratio of TS to reactant partition functions**

TST involves the ratio:

$$
\frac{Q^\ddagger}{Q_\text{reactants}} 
= \frac{ e^{-E_\text{TS}/(k_B T)} \sum_i e^{-(E_i^\ddagger - E_\text{TS})/(k_B T)} }
       { e^{-E_\text{R}/(k_B T)} \sum_j e^{-(E_j^\text{react} - E_\text{R})/(k_B T)} }
$$

Factor out the minimal energies:

$$
\frac{Q^\ddagger}{Q_\text{reactants}} 
= e^{-(E_\text{TS} - E_\text{R})/(k_B T)} 
\cdot 
\frac{\sum_i e^{-(E_i^\ddagger - E_\text{TS})/(k_B T)}}{\sum_j e^{-(E_j^\text{react} - E_\text{R})/(k_B T)}}
$$

* **Step 4 — Identifying the exponential factor**

Define the **activation energy**:

$$
E_0 = E_\text{TS} - E_\text{R}
$$

Then the ratio becomes:

$$
\frac{Q^\ddagger}{Q_\text{reactants}} = e^{-E_0/(k_B T)} \cdot \frac{Q^\ddagger_\text{non-elec}}{Q_\text{reactants, non-elec}}
$$

where  
- $Q^\ddagger_\text{non-elec}$ = all but electronic contributions to the partition function of TS (excluding vibrational reaction coordinate)
- $Q_\text{reactants, non-elec}$ = all but electronic contributions to the partition function of reactants

* **Step 5 — Full TST rate constant**

Finally, including motion along the reaction coordinate:

$$
\boxed{
k(T) = \frac{k_B T}{h} \frac{Q^\ddagger_\text{vib}}{Q_\text{reactants, vib}} \, e^{-E_0/(k_B T)}
}
$$

- The **exponential factor** $e^{-E_0/(k_B T)}$ comes from the **Boltzmann probability** of molecules reaching the TS.  
- The remaining partition function ratio gives the **pre-exponential factor**.



## 1.4. True vs Experimental Activation Energy

In chemical kinetics, it is important to distinguish between **true (intrinsic)** and **experimental (observed)** activation energies.

### A. True (intrinsic) activation energy

- Denoted \($E_a^{\rm true}$\) or \($E_a^0$\).  
- Represents the **energy barrier of the elementary reaction**.  
- It is the **potential energy difference** between reactants and the transition state along the reaction coordinate.  
- This is what is used in collision theory or transition-state theory:

$$
k(T) = p\, Z \, e^{-E_a^{\rm true}/(RT)},
$$

where \(p\) is the steric factor and \(Z\) is the collision frequency.

### B. Experimental (observed) activation energy

- Denoted \($E_A^{\rm exp}$\).  
- Measured from an **Arrhenius plot**:

$$
\ln k \text{ vs } 1/T \quad\implies\quad \text{slope} = -\frac{E_A^{\rm exp}}{R}.
$$

- Includes **all real-world effects**, not just the intrinsic barrier:
  1. Orientation / steric effects \($p<1$\)  
  2. Diffusion or collision frequency limitations  
  3. Solvent effects (for solution reactions)  
  4. Pre-equilibrium or complex formation
  
- Relation to the observed rate constant:

$$
k_{\rm obs}(T) = k_0\, e^{-E_A^{\rm exp}/(RT)}.
$$


### C. Relation between true and experimental activation energies

$$
k_{\rm obs} = p\, Z \, e^{-E_a^{\rm true}/(RT)} = A_{\rm exp}\, e^{-E_A^{\rm exp}/(RT)}.
$$

Taking logarithms:

$$
\ln k_{\rm obs} = \ln(p Z) - \frac{E_a^{\rm true}}{RT} = \ln A_{\rm exp} - \frac{E_A^{\rm exp}}{RT}.
$$

- If \(p\) is **temperature-independent**, then \($E_A^{\rm exp} = E_a^{\rm true}$\).  
- If \(p\) depends on \(T\):

$$
E_A^{\rm exp} = E_a^{\rm true} + R \frac{d \ln p}{d (1/T)}.
$$

In summary:

| Feature | True activation energy | Experimental activation energy |
|---------|----------------------|--------------------------------|
| Definition | Energy barrier of the elementary reaction | Extracted from Arrhenius plot of \(k(T)\) |
| Includes steric factors? | No | Yes |
| Includes diffusion / solvent effects? | No | Yes, if measured in experiment |
| How measured | Computed from potential energy surface / transition state | Derived from slope of \(ln k\) vs \(1/T\) |

## Problem 1

Compute the steric factor for the ethylene dimerization reaction

$2C_2H_4 \rightarrow C_4H_8$

at T = 300K, if the observed activation energy is $E_A = 146.5 kJ mol^{-1}$. Assume the effective diameter of ethylene is 0.49 nm and the experimental rate constant at this temperature is $k = 1.08 \times 10^{-14} cm^3 mol^{-1} s^{-1} $

### Solution

```python
import math
import numpy as np


M = (2*12 + 4*1)
mu = M/2
print(F"mu = {mu} g/mol") 

# we need to use SI units, so kg/mol
mu = mu * 0.001

# 0.49 nm = 4.9*10^-10 m
d = 4.9e-10 # m

Na = 6.023e+23 # Avogadro's number

k = 1.08e-14 # cm^3/mol*s

# convert cm^3 to m^3    1 cm^3 = 10^{-6} cm^3
k = k * 1e-6 # no in m^3/mol*s

T = 300  # K

# Use R instead of k_B, since we are working with the molar quantities
R = 8.314

#### NEEED ATTENTION !!! ##########
# True activation energy from the experimental one:
# for bimolecular reactions, Ea = EA - 1/2 RT
EA = 146.5e+3 # J/mol
Ea = EA - 0.5*R*T

# theoretical rate constant:
k_theor = Na* d**2 * math.sqrt( 8 * math.pi * R * T/mu) * math.exp(-Ea/(R*T))

print(F"Experimental k = {k}")
print(F"Computed k = {k_theor}")
print(F"Steric factor = {k/k_theor:.2e}")
```
