# Significance

* **Ideal gas law ($PV=nRT$)**
– Foundation for stoichiometric gas calculations in reactions, yields, and volumetric gas analysis.

* **Real gas behavior & equations of state (van der Waals, Berthelot, Redlich–Kwong, etc.)**
– Important for high-pressure reactions, supercritical fluids, and industrial synthesis (e.g., ammonia in the Haber process).

* **Gas mixtures & Dalton’s law of partial pressures**
– Relevant in combustion, environmental chemistry (pollutants in air), and chromatography.

* **Respiratory physiology**
– Gas laws explain lung ventilation, oxygen uptake, and CO₂ removal. Dalton’s law governs alveolar gas exchange.

* **Hyperbaric and hypobaric medicine**
– Treatment of decompression sickness, wound healing, and oxygen therapy depends on non-ideal gas behavior under pressure.

* **Medical devices**
– Gas laws underpin the operation of ventilators, anesthesia vaporizers, and pressurized inhalers.

* **Altitude and diving physiology**
– Boyle’s law and Henry’s law explain effects of high altitude (hypoxia) and diving (nitrogen narcosis, “the bends”).

## 1.1. Basic definitions 

Thermodynamics is the branch of physics that deals with the relationships between heat, work, temperature, and energy. It describes how energy is transferred between physical systems and how it affects matter. Here are some fundamental terms and definitions:

1. **System:** A specific portion of the universe that is under study. It can be as small as a gas in a container or as large as an entire planet.
   
   - **Isolated System:** A system that does not exchange energy or matter with its surroundings.
   - **Closed System:** A system that can exchange energy (heat and work) but not matter with its surroundings.
   - **Open System:** A system that can exchange both energy and matter with its surroundings.
   

2. **Surroundings (Environment):** Everything outside the system that can interact with it.

4. **State Functions:** Properties that depend only on the current state of the system, not on how it got there.

    Examples include internal energy (U), enthalpy (H), entropy (S), and pressure (P).

   - **Internal Energy (U):** The total energy contained within a system, which includes kinetic and potential energy at the molecular level.

   - **Enthalpy (H):** A measure of the total energy of a thermodynamic system, including internal energy and the product of its pressure and volume.
    It is defined as $H = U + PV$.

   - **Entropy (S):** A measure of the randomness or disorder of a system. It is a state function that increases with the number of microstates of a system.
  
5. **Path functions:** Functions that descrbe the change of properties of a thermodynamic system that depend in the path of going from one state to another state. The main examples are:  heat (Q) and work (W)

   - **Heat(Q):** A form of energy transfer due to temperature difference.

   - **Work (W):** Energy transfer associated with a force acting through a distance.

6. **Process:** A transformation from one state to another. Processes can be:

   - **Isothermal:** Constant temperature.
   - **Adiabatic (isoentropic):** No heat exchange with the surroundings.
   - **Isobaric:** Constant pressure.
   - **Isochoric:** Constant volume.


## 1.2. Gas Laws

Gas laws describe the behavior of gases in terms of temperature, pressure, and volume. Some of the fundamental gas laws are:

1. **Boyle's Law:** At constant temperature, the pressure of a gas is inversely proportional to its volume.
    $P ~ \frac{1}{V}$ or $PV=constant$
   
    
2. **Charles's Law:** At constant pressure, the volume of a gas is directly proportional to its absolute temperature.

    $V ~ T$ or $\frac{V}{T}=constant$

3. **Gay-Lussac's Law:** At constant volume, the pressure of a gas is directly proportional to its absolute temperature.

   $P ~ T$ or $\frac{P}{T}=constant$

4. **Avogadro's Law:** At constant temperature and pressure, the volume of a gas is directly proportional to the number of moles of gas.

   $V ~ n$ or $\frac{V}{n}=constant$

5. **Ideal Gas Law:** A combination of Boyle's, Charles's, and Avogadro's laws. It describes the behavior of an ideal gas.

   $PV=nRT$
   where:
   - P = Pressure
   - V = Volume
   - n = Number of moles
   - R = Universal gas constant (8.314 J/(mol * K)
   - T = Absolute temperature in Kelvin

## 1.3. Equations of State for Gases

An equation of state is a mathematical model describing the state properties of a gas. The most commonly used equations of state for gases are:

1. **Ideal Gas Equation:**

   $PV=nRT$
    
   Assumes no interactions between gas molecules and that the volume occupied by the gas molecules themselves is negligible.

2. **Van der Waals Equation:** A more realistic equation for real gases that takes into account intermolecular
   forces and the finite volume of gas molecules:

   $(P+\frac{a}{V_m^2} )(V_m−b) = nRT$

    where:
    - $a$ and $b$ are empirical constants specific to each gas, accounting for intermolecular attraction and finite molecular volume, respectively.
    - $V_m = \frac{V}{n}$ - molar volume of gas

3. **Dieterici Equation of State:** The Dieterici equation of state accounts for intermolecular attraction and is another
  modification of the ideal gas law, similar to the Van der Waals equation. The Dieterici equation includes an exponential term
  that reduces the pressure compared to the ideal gas law, incorporating both repulsive and attractive forces among molecules more effectively.
  It is given by:

   $P = \frac{nRT}{V − nb}exp⁡(−\frac{a}{RTV})$
   where:

   - a = Attraction parameter (accounts for intermolecular forces)
   - b = Volume correction parameter (accounts for finite molecular volume)

4. **Berthelot Equation of State:** The Berthelot equation of state is another modification of the ideal gas law that introduces corrections for intermolecular forces and the finite size of gas molecules. It is particularly useful at moderate pressures and temperatures close to the critical point.
The Berthelot equation is a modification of the Van der Waals equation, where the attraction term is inversely proportional to temperature. This form provides a better approximation for gases near their critical points than the ideal gas law or simpler equations. The Berthelot equation is given by:

    $P=\frac{nRT}{V−nb}−\frac{n^2}{aTV^2}$

5. **Virial Equation of State:** An equation that expands the ideal gas law to account for real gas behavior using a power series in terms of $\frac{1}{V}$:
    $PV=nRT (1 + \frac{B_1(T)}{V} + \frac{B_2(T)}{V^2} +⋯ )$

    where $B_1(T)$, $B_2(T)$, etc. are first, second and so on virial coefficients that depend on temperature and the type of gas.

## Problem 1

Atkins 1.3

Much to everyone’s surprise, nitrogen monoxide (NO) has been found to act as a neurotransmitter. To prepare to study its effect, a sample was collected in a container of volume 300.0 $cm^3$. At 14.5°C its pressure is found to be 34.5 kPa. What amount (in moles) of NO has been collected?

### Solution

```python
cm3_to_m3 = 1.0e-6 # conversion from cm^3 to m^3 

V = 300.0 # cm^3
V = V * cm3_to_m3 # m^3 
T = 14.5 + 273.15 # T in K
P = 34500.0 # Pa 
R = 8.314 # J/mol*K

# PV = nRT
n = P*V/(R*T)

print(n)
```

## Problem 2

Semiokhin 5

Compute the pressure of 2 moles of chlorobenzene occupying volume of 10 L at 25 Celsius using: a) the ideal gas law; b) the van der Waals equation. The van der Waals equation a and b parameters for chlorobenzene are $a = 25.43 L^2 \times atm \times mol^{-2}$; $b = 0.1453 L \times mol^{-1}$

### Solution

```python
L_to_m3 = 1e-3 
n = 2.0 # moles
V = 10.0 # L
T = 25.0 + 273.15 # K

a = 25.43 # L^2 * atm / mol^2
b = 0.1453 # L/mol
R = 0.0821 # L*atm/(mol*K)

P = n * R * T/V # atm

print(F"P of the ideal gas = {P}")  # in atm


# (P + a/V_m^2) * (V_m - b) = RT
# (P + a/V_m^2)  = RT / (V_m - b)
# P   = RT / (V_m - b) - a/V_m^2
Vm = V / n 

P = R*T / (Vm - b) - a/Vm**2
print(F"P of the vdW gas = {P}")
```

## Problem 3

Atkins 1.15

A gas mixture being used to simulate the atmosphere of another planet consists of 320 mg of methane, 175 mg of argon, and 225 mg of nitrogen. The partial pressure of nitrogen at 300 K is 15.2 kPa. Calculate (a) the volume and (b) the total pressure of the mixture.

### Solution

```python
m_ch4 = 320 # mg 
m_ar = 175 # mg
m_n2 = 225 # mg

# convert to g
m_ch4 = m_ch4 * 1e-3
m_ar = m_ar * 1e-3
m_n2 = m_n2 * 1e-3

M_ch4 = 12 + 4*1 # g/mol
M_ar = 40 # g/mol
M_n2 = 14 * 2  # g/mol
T = 300.0 # K
p_n2 = 15.2 # kPa

# Solution strategy
# PV = nRT 
# P_n2 = x_n2 * P

# Let's compute the number of moles and then molar fractions of each component
n_ch4 = m_ch4 / M_ch4
n_ar = m_ar / M_ar
n_n2 = m_n2 / M_n2
n = n_ch4 + n_ar + n_n2  # the total number of moles of all gases 

# Just need the molar fraction of N2
x_n2 = n_n2 / n

# Then solve for the total pressure of the mixture
P = p_n2 / x_n2

print(F"Total pressure = {P} kPa")

# Now solve for the volume - need to convert the pressure to
# the SI units - Pa, instead of kPa:

P = P * 1000.0 # P in Pa
R = 8.314 # J /mol*K
V = n * R * T/ P # m^3
print(V)
```

## Problem 4

Compute the density of an air saturated with water vapor at T=25℃. The partial pressure of water vapor under these conditions is 23.7 mmHg. The total pressure is 1 atm. The composition of the water-vapor-free air is 78.1% of N2, 21% of O2, and 0.9% of Ar. The percentage given is the composition by volume. Report your answer in units of g/L. Round to third decimal place. 

### Solution

See below

```python
# Constants
mmHg2atm = 1.0/760.0  # P[atm] = P[mmHg] * mmHg2atm
R = 0.0821 # L*atm/mol*K
atm2Pa = 101325.0 #  P[Pa] = P[atm] * atm2Pa
```

```python
# First determine the composition of the system
p_vap = 23.7 * mmHg2atm # partial pressure of water wapor in atm
p_tot = 1.0 # total pressure of the mixture, atm 

x_vap = p_vap / p_tot # molar fraction of water vapor
x_air = 1.0 - x_vap   # molar fraction of the dry air = mixture of N2, O2, and Ar

# Now determine the absolute percentages of N2, O2, and Ar in the mixture
x_n2 = 0.781 * x_air 
x_o2 = 0.21 * x_air
x_ar = 0.009 * x_air

# Now, assume there is 1 mole of the gas mixture. What's it volume under such conditions?
# we use the ideal gas law: PV = nRT, n = 1
T = 25 + 273.15  # convert T to K
n = 1.0 # mole
V = n * R * T / p_tot  # volume in L
print(F"volume = {V}")

# Now compute the total mass of 1 mole of the mixture - we have to use molar masses
M_h2o = 18.0 # molar mass of water, g/mol 
M_o2 = 16 * 2 # O2
M_n2 = 14 * 2 # N2
M_ar = 40.0   # Ar

mass = M_h2o*x_vap + M_o2*x_o2 + M_n2*x_n2 + M_ar*x_ar  # mass in g

dens = mass / V # g/L, but it is more convenient to use g/ml

print(F"density = {dens}")
```

## Problem 5

In the first step, each of two spherical flasks (22.4 L each) is filled with 1 mole of gaseous nitrogen (per flask) at T=25℃ and P=1 atm. In the second step, the two flasks are connected by a small tube of negligible volume. Finally, each of the flasks is placed in its own bath of different temperature: one is thermostatted at 100℃ and the other one – at 0 ℃. Determine the final pressure (in atm) and the final number of moles of nitrogen in each flask. 

### Solution

The pressure in the first flask will be $P_1 = \frac{n_1 R T_1}{V_1}$. The pressure of N2 in the second flask will be:
$P_2 = \frac{n_2 R T_2}{V_2}$. 

The two flasks are in equilibrium, so the pressures should be equal: $P_1 = P_2$. Because $V_1 = V_2$, we have:

$n_1 T_1 = n_2 T_2$

We don't know the number of moles of N2 in each flask, but we know that the total number of moles of N2 placed in the two flasks is 2 moles, so:

$n_1 + n_2 = 2.0$

In code cell, we solve this problem and then we can compute the pressure:

```python
# n1 + n2 = 2.0
# n1/n2 = T2/T1
T1 = 100 + 273.15
T2 = 0.0 + 273.15

n1_to_n2 = T2/T1

#n2 * (n1_to_n2 + 1) = 2.0
# so:
n2 = 2.0 / (1 + n1_to_n2)  # number of moles of N2 in the cold flask
n1 = 2.0 - n2              # number of moles of N2 in the hot flask

V1 = 22.4
P1 = n1 * R * T1 / V1      # pressure in both flasks in atm

print(F"moles in cold flask = {n2} moles in hot flask = {n1} pressure in each flask = {P1} atm")
```
