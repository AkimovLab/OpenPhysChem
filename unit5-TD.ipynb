# 1. Theory

## 1.1. Entropy (S)

### A. Definitions

- **Microscopic (Boltzmann definition)**:

$$
S = k_B \ln W
$$

where:
- $W$ = number of accessible microstates consistent with the macroscopic state,  
- $k_B$ = Boltzmann constant.  

- **Thermodynamic definition**:

$$
\Delta S = \frac{\delta q_\text{rev}}{T}
$$

### B. Explanations

- Matter and energy tend to disperse spontaneously
- To measure the degree of matter/energy dispersion, we use term **entropy**
- Entropy is a **measure of disorder** in a system
- In definition: Heat not work, because it is related to disorderly motion (heat)
- In definition: reversible ("rev") – to have the equality
- In definition: T in the denominator – same $\delta 𝑄$ causes a higher relative disorder at lower temperatures

- **Spontaneous** - having a natural tendency to occur without work to be done
- **Non-spontaneous** – can only be realized by doing work

### C. Statistical origins of entropy

- Entropy measures **probability**: more microstates → higher entropy.  
- Spontaneous processes move toward macroscopic states that are realized by **larger number of microscopic states** (larger $W$).  


## 1.2. Second Law of Thermodynamics

**First form:**  
> In any real process, the total entropy of the universe (system + surroundings) never decreases:

$$
\Delta S_\text{universe} = \Delta S_\text{system} + \Delta S_\text{surroundings} \geq 0
$$

- Reversible:  $\Delta S_\text{universe} = 0$  
- Irreversible (spontaneous):  $\Delta S_\text{universe} > 0$  

**Second form:**
> There exists a state function, **S (called entropy)**, such that:
> 1) $dS=\frac{\delta Q_\text{rev}}{T}$ for reversible/equilibrium processes
> 2) $dS > \frac{\delta Q_\text{irrev}}{T}$ for irreversible/non-equilibrium processes

**Third form:**
> Perpetuum mobile of the 2nd kind doesn’t exist

**Definition:** Perpetuum mobile of the 2-nd kind: converts heat extracted from all surroundings into a useful work.
(it doesn't violate the 1-st law, but violates the 2-nd one)


## 1.3. Carnot Cycle and the Second Law

### A. Definitions

The **Carnot cycle** is an idealized, reversible heat engine cycle consisting of:
1. Isothermal expansion at high temperature $T_H$  
2. Adiabatic expansion (temperature drops from $T_H$ to $T_C$)  
3. Isothermal compression at low temperature $T_C$  
4. Adiabatic compression (temperature rises from $T_C$ back to $T_H$)  

**Significance:** a formal way to arrive at the TD definition of entropy and show it is a state function.

### B. Efficiency of a Carnot Engine

The thermal efficiency is:

$$
\eta = 1 - \frac{Q_C}{Q_H}
$$

For a reversible Carnot engine:

$$
\frac{Q_C}{Q_H} = \frac{T_C}{T_H}
$$

So:

$$
\eta = 1 - \frac{T_C}{T_H}
$$



## 1.4. Practical Formulas for Computing Entropy Changes


### A. Isothermal Reversible Heat Transfer

For a reversible process:

$$
\Delta S = \int \frac{\delta q_\text{rev}}{T}
$$

### B. Isothermal Expansion/Compression of an Ideal Gas

For $n$ moles:

$$
\Delta S = nR \ln \frac{V_f}{V_i} = nR \ln \frac{P_i}{P_f}
$$

(since $T=\text{const}$, and $PV=nRT$).

### C. Temperature Change at Constant Heat Capacity

If heat capacity $C$ is constant:

$$
\Delta S = C \ln \frac{T_f}{T_i}
$$

- At constant pressure: $C = C_p$  
- At constant volume: $C = C_v$  


### D. Phase Transitions

At temperature $T_\text{trans}$:

$$
\Delta S = \frac{\Delta H_\text{trans}}{T_\text{trans}}
$$

Examples:
- Fusion: $\Delta S_\text{fus} = \frac{\Delta H_\text{fus}}{T_m}$  
- Vaporization: $\Delta S_\text{vap} = \frac{\Delta H_\text{vap}}{T_b}$  


## 1.5. Free Energies and Spontaneity

Entropy change of the universe can be reformulated in terms of **free energies**.

### A. Helmholtz Free Energy (F or A)

$$
F = U - TS
$$

- Useful at **constant T and V**.  
- Spontaneity criterion:

$$
\Delta F \leq 0 \quad (T,V=\text{const})
$$

### B. Gibbs Free Energy (G)

$$
G = H - TS
$$

with enthalpy

$$
H = U + PV
$$

- Useful at **constant T and P**.  
- Spontaneity criterion:

$$
\Delta G \leq 0 \quad (T,P=\text{const})
$$

- Relation to entropy of the universe:

$$
\Delta G = -T \Delta S_\text{universe}
$$

Thus:

- If $\Delta G < 0$ → spontaneous  
- If $\Delta G = 0$ → equilibrium  
- If $\Delta G > 0$ → non-spontaneous  

### C. Spontaneous Processes and Directionality

- **Spontaneous:** proceeds without external input once initiated.  
- **Directionality:** determined by the sign of $\Delta G$ (at constant T, P) or equivalently $\Delta S_\text{universe}$.  

## 1.6. Third Law of Thermodynamics

### A. Formulation

**Third Law Statement:**
> The entropy of a perfect crystal at absolute zero temperature is zero:

$$
S(0 \,\text{K}) = 0
$$

(provided the crystal has a unique ground state).


### B. Absolute Entropy

From heat capacity and phase transition data:

$$
S(T) = \int_0^T \frac{C_p}{T}\, dT \; + \; \sum \frac{\Delta H_\text{trans}}{T_\text{trans}}
$$

- Absolute entropies are tabulated (e.g., at 298 K).


### C. Residual Entropy

- $T = 0$ no vibrations
- Ideal geometry – no distortions

But computed and experimental entropies are different!

- For 𝐶𝑂 the difference is $6\ J\cdot K^{−1}\cdot mol^{−1}$
- The CO molecules are disordered in the crystal

Assume each can be in 2 orientations
The number of configurations $W = 2^{N_A}$

Entropy correction is:
$$
𝑆_{𝑐𝑜𝑟𝑟}=𝑘_𝐵\ ln⁡(𝑊) = 𝑘_𝐵 ln⁡(2^{𝑁_𝐴} ) = 𝑁_𝐴\ 𝑘_𝐵\ ln⁡(2) = 𝑅\ ln⁡(2) = 5.8\ J\cdot K^{−1}\cdot mol^{−1}
$$

## Problem 1 [Eremin Example 4-1]

Determine the dependence of entropy on volume for a thermodynamic system that described by the equation of the state (for one mole)

$(p + \frac{a}{V^2})(V - b) = RT$

### Solution

Solve for P:

$P = \frac{RT}{V-b} - \frac{a}{V^2}$

Compute the derivative:

$(\frac{\partial P}{\partial T})_V = \frac{R}{V-b}$

The entropy change is computed as:

$\Delta S = \int_{V_1}^{V_2} (\frac{\partial P}{\partial T})_V dV = \int_{V_1}^{V_2} \frac{R}{V-b} dV = R ln(\frac{V_2-b}{V_1-b}) $

## Problem 2 [Eremin Example 4-2]

Calculate the change in entropy when 0.7 mole of monoclinic sulfur is heated from 25 to 200 Celsius at 1 atm pressure. Molar heat capacity of sulfur is given by:

$C_{p,m}(S(s)) = 23.64$ &nbsp; $J \times mol^{-1}K^{-1}$,

$C_{p,m}(S(l)) = 35.73 + 1.17 \times 10^{-3} T$ &nbsp; $J \times mol^{-1}K^{-1}$

The melting temperature of monoclinic sulfur is 119 Celsius, the specific heat of melting 45.2 J/g

### Solution

The total change of entropy $\Delta S$ is composed on entropy changes of three processes:

* heating the solid from $T_1 = 25$ Celsius to the melting point $T_2 = 119$ Celsius, $\Delta S_1 = n \int_{T_1}^{T_2} \frac{C_{p,m}(s)}{T} dT$

because the solid sulfur's heat capacity is constants, the integral becomes:

$n \int_{T_1}^{T_2} \frac{C_{p,m}(s)}{T} dT = n C_{p,m}(s) ln(\frac{T_2}{T_1})$

* melting the solid sulfur at $T_2 = 119 C^\circ$

$\Delta S_2 = m \frac{Q_{s}}{T_2} $

* heating the liquid from the melting point $T_2 = 119$ Celsius to the desired temperature $T_3 = 200$ Celsius

In this case, heat capacity depends on temperature: $C_{p,m}(l) = a + b T$ so

$\Delta S_3 = n \int_{T_2}^{T_3} \frac{C_{p,m}(l)}{T} dT = a ln(\frac{T_3}{T_2}) + b(T_3 - T_2)$

```python
import math 

n = 0.7 # moles
T1 = 25 + 273.15 # K
T2 = 119 + 273.15 # K melting temperature
T3 = 200 + 273.15 # K
Cpm_solid = 23.64 # J / mol * K 
Q_melt = 45.2 # J/g  melting heat 
m = 32.0 * n # molar mass of sulfur

a = 35.73 
b = 1.17e-3 

dS1 = n * Cpm_solid * math.log(T2 / T1)
dH = m * Q_melt 

dS2 = dH / T2 

dS3 = n * ( a * math.log(T3/T2) + b * (T3 - T2) )

dS = dS1 + dS2 + dS3

print(F"dS = {dS} J/K")
```

## Problem 4

(5 points max) Compute $\Delta S$ for heating 0.4 mole of NaCl from 20℃ to 850℃. The molar heat capacity of NaCl in solid and liquid states is given by: 

$C_{p,m}(NaCl,solid)=45.94 + 16.32*10^{-3} T  \frac{J}{(mol*K}$

$C_{p,m}(NaCl,liquid)=66.53  \frac{J}{mol*K}$

NaCl melts at 800 ℃ with the molar fusion heat of $\Delta_{fus} H = 31.0 kJ/mol$

### Solution

The process consists of 3 steps - heating up the solid until the fusion temperature, melting, and then heating the liquid phase up to the target temperature.

The first process:

$\Delta S_1 = n \int_{T_1}^{T_2} \frac{C_{p,m}(solid)}{T} dT = n \int_{T_1}^{T_2} \frac{45.94 + 16.32*10^{-3} T}{T} dT = 45.94 ln(\frac{T_2}{T_1}) + 16.32*10^{-3} * (T_2 - T_1)$

The second process:

$\Delta S_2 = n \frac{\Delta_{fus} H}{T_{fus}} = n \frac{\Delta_{fus} H}{T_{2}}$

The third process:

$\Delta S_3 = n \int_{T_2}^{T_3} \frac{C_{p,m}(liquid)}{T} dT = n \int_{T_2}^{T_3} \frac{66.53}{T} dT = 66.53 ln(\frac{T_3}{T_2})$

```python
T1 = 20 + 273.15
T2 = 800 + 273.15
T3 = 850 + 273.15
n = 0.4 # moles of NaCl

dH_fus = 31.0e+3 # J/mol

dS1 = n * (45.94 * math.log(T2/T1) + 16.32e-3 * (T2 - T1) )

dS2 = n* dH_fus / T2

dS3 = n * 66.53 * math.log(T3/T2)

dS = dS1 + dS2 + dS3

print(F" dS = {dS} J/K")
```

## Problem 5

(5 points max). An iron (Fe) block of mass 0.5 kg heated to temperature t=100℃ is immersed into 1 kg of water at t=0℃. Compute $\Delta S_{Fe}$, $\Delta S_{water}$, and the $\Delta S_{tot}$ (where “tot” means the Fe block + water), if the specific heat capacities of iron and water are $C_{Fe} =0.447 \frac{J}{g K}$ and $C_{water} = 4.184 \frac{J}{g K}$ respectively. We assume that the iron + water system is isolated. Does your result agree with an expectation based on the second law of thermodynamics. What is that expectation? Explain. 

### Solution

First, we need to compute determine the amount of heat transfered between components (Fe block and water).

Heat transferred from Fe block will be:

$|q_{Fe}| = m_{Fe} C_{Fe} (T_{Fe} - T_{eq})$

where $T_{Fe} = 100 + 273.15 K$ and $T_{eq}$ is yet unknown temperature at equilibrium.

Likewise, heat transferred to water will be:

$|q_{water}| = m_{water} C_{water} (T_{eq} - T_{water})$

At equilibrium,

$|q_{Fe}| = |q_{water}|$

So:

$m_{Fe} C_{Fe} (T_{Fe} - T_{eq}) = m_{water} C_{water} (T_{eq} - T_{water})$

So:

$T_{eq} = \frac{m_{Fe} C_{Fe} T_{Fe} + m_{water} C_{water} T_{water}}{m_{Fe} C_{Fe} + m_{water} C_{water}} $

```python
T_Fe = 100 + 273.15 # K
T_water = 0.0 + 273.15 # K
C_Fe = 0.447 # J/g*K
C_water = 4.184 # J/g*K
m_Fe = 0.5 * 1000.0 # g
m_water = 1.0 * 1000 # g

# Compute the equilibrium temperrature
T_eq = (m_Fe * C_Fe * T_Fe + m_water * C_water * T_water)/(m_Fe * C_Fe + m_water * C_water)

print(F" equilibrium temperature = {T_eq - 273.15} Celsius")

# Now compute the amount of heat transferred:
q_water = m_water * C_water * (T_eq - T_water)
print(F" heat trasferred to water = {q_water} J")

q_Fe = -q_water
print(F" heat trasferred from Fe = {q_Fe} J")


dS_Fe = q_Fe/T_Fe
print(F" dS for Fe = {dS_Fe} J/K")

dS_water = q_water/T_water
print(F" dS for water = {dS_water} J/K")

dS_tot = dS_Fe + dS_water
print(F" dS total = {dS_tot} J/K")
```

### Answers 

$\Delta S_{Fe} = -56.9 $ J/K

$\Delta S_{water} = 77.7 $ J/K

$\Delta S_{tot} = 20.8 $ J/K

The overall change of entropy is positive, as expected from the second law of TD for an isolated system. So, the heat transfer from Fe to water occurs spontaneously.
