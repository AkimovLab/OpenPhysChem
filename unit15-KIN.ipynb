## 1. Theory

## 1.1. Kinetics of complex (composite) reactions

### A. Consecutive (Sequential) Reactions
$$
A \xrightarrow{k_1} B \xrightarrow{k_2} C
$$

Differential equations:
$$
\begin{aligned}
\frac{d[A]}{dt} &= -k_1[A] \\
\frac{d[B]}{dt} &= k_1[A] - k_2[B] \\
\frac{d[C]}{dt} &= k_2[B]
\end{aligned}
$$

Integrated solutions (for $A(0) = A_0$, $B(0) = 0$):
$$
\boxed{
\begin{aligned}
{[A]}(t) &= A_0 e^{-k_1 t} \\
[B](t) &= \frac{A_0 k_1}{k_2 - k_1} \left(e^{-k_1 t} - e^{-k_2 t}\right) \\
[C](t) &= A_0 \left[1 - \frac{k_2 e^{-k_1 t} - k_1 e^{-k_2 t}}{k_2 - k_1}\right] = \frac{k_2\!\left(1 - e^{-k_1 t}\right) - k_1\!\left(1 - e^{-k_2 t}\right)}{k_2 - k_1}\,[A]_0
\end{aligned}
}
$$
→ $B(t)$ first increases, then decays.

### B. Parallel (Competitive) Reactions
$$
A \xrightarrow{k_1} B, \quad A \xrightarrow{k_2} C
$$

Rate of A:
$$
\frac{d[A]}{dt} = -(k_1 + k_2)[A]
$$

Integrated:
$$
\boxed{
\begin{aligned}
{[A]}(t) = [A]_0 e^{-(k_1 + k_2)t}\\
[B](t) = [A]_0 \frac{k_1}{k_1 + k_2} \left(1 - e^{-(k_1 + k_2)t}\right) \\
[C](t) = [A]_0 \frac{k_2}{k_1 + k_2} \left(1 - e^{-(k_1 + k_2)t}\right)
\end{aligned}
}
$$

### C. Reversible Reaction
$$
A \overset{k_1}{\underset{k_{-1}}{\rightleftharpoons}} B
$$

Differential equation:
$$
\frac{d[A]}{dt} = -k_1[A] + k_{-1}[B]
$$

Let total concentration be constant: $[A] + [B] = [A]_0 + [B]_0 = C_T$.

Integrated form:
$$
\boxed{
[A](t) = [A]_\text{eq} + ([A]_0 - [A]_\text{eq}) e^{-(k_1 + k_{-1})t} \\
[A]_\text{eq} = \frac{k_{-1} C_T}{k_1 + k_{-1}}, \quad [B]_\text{eq} = \frac{k_1 C_T}{k_1 + k_{-1}}
}
$$

Alternatively, in terms of **degree of conversion**, $x$:  $[A] = [A]_0 - x$ and $[B] = [B]_0 + x$

At equilibrium:
$$
\frac{dx}{dt} (x=x_{eq}) = k_1 ([A_0] - x_{eq}) - k_{-1} ([B]_0 + x_{eq}) = 0 \Rightarrow 
K = \frac{[B]_{eq}}{[A]_{eq}} = \frac{[B]_0 + x_{eq}}{[A]_0 - x_{eq}} = \frac{k_1}{k_{-1}}
$$

So:
$$
\boxed{
x(t) = x_{eq} \left[1 - \exp(-(k_1 + k_{-1})t ) \right] \\
x_{eq} = \frac{k1 [A]_0 - k_{-1} [B]_0} {k_1 + k_{-1}}
}
$$


## 1.2. Determination of the reaction orders

The **order of a reaction** describes how the **rate** depends on the **concentration of reactants**.

For a general rate law:

$$
r = k [A]^m [B]^n
$$

- m and n are the **partial reaction orders** with respect to A and B.
- The **overall order** is m + n.
- Orders are determined **experimentally** — not from stoichiometry (except for elementary steps).

### A. Differential (Initial Rate) Method

Measure the **initial rate** $r_0$ for different initial concentrations.

For reaction:
$$
A + B \rightarrow \text{Products}
$$

Rate law:
$$
r_0 = k [A]_0^m [B]_0^n
$$

Take logarithms:
$$
\ln r_0 = \ln k + m \ln [A]_0 + n \ln [B]_0
$$

#### A.1 Isolate one reactant
Keep $[B]_0$ constant, vary $[A]_0$:
$$
\ln r_0 = m \ln [A]_0 + \text{const.}
$$
→ Slope of plot $\ln r_0 $ vs. $\ln [A]_0 $ gives **m**.

Repeat with $[A]$ constant to obtain **n**.


#### A.2 Example
| Exp | [A]₀ (mol/L) | [B]₀ (mol/L) | r₀ (mol L⁻¹ s⁻¹) |
|-----|---------------|---------------|-------------------|
| 1   | 0.10 | 0.10 | 1.5×10⁻³ |
| 2   | 0.20 | 0.10 | 6.0×10⁻³ |
| 3   | 0.20 | 0.20 | 1.2×10⁻² |

From Exps 1 and 2 (B constant):
$$
\frac{r_2}{r_1} = \left(\frac{[A]_2}{[A]_1}\right)^m \Rightarrow \frac{6.0\times10^{-3}}{1.5\times10^{-3}} = 4 = 2^m \Rightarrow m = 2
$$

From Exps 2 and 3 (A constant):
$$
\frac{r_3}{r_2} = \left(\frac{[B]_3}{[B]_2}\right)^n \Rightarrow 2 = 2^n \Rightarrow n = 1
$$

Thus:  
**Rate law:** $r = k [A]^2 [B]$

### 2. Integrated Rate Law Method

For simple reactions, plot concentration vs. time and compare to integrated forms.

| Order | Differential form | Integrated form | Test plot | Linear relation |
|-------|--------------------|------------------|------------|----------------|
| 0 | $$\frac{d[A]}{dt} = -k $$ | $$[A] = [A]_0 - kt $$ | $$[A]\ \text{vs.} t$$ | slope = –k |
| 1 | $$\frac{d[A]}{dt} = -k[A]$$ | $$ \ln[A] = \ln[A]_0 - kt $$ | $$\ln[A]\ \text{vs.} t$$ | slope = –k |
| 2 | $$\frac{d[A]}{dt} = -k[A]^2$$ | $$ \frac{1}{[A]} = \frac{1}{[A]_0} + kt $$ | $$1/[A]\ \text{vs.} t$$ | slope = k |

By testing these plots, the one that gives a straight line identifies the **reaction order**.

### 3. Half-Life Method

The **half-life** $\tau_{1/2}$ is the time for concentration to drop to half its initial value.

| Order | Expression for $$\tau_{1/2}$$ | Dependence on $[A]_0$ |
|-------|-------------------------------|---------------------|
| 0 | $$\tau_{1/2} = \frac{[A]_0}{2k} $$ | proportional to $[A]_0$ |
| 1 | $$\tau_{1/2} = \frac{\ln 2}{k} $$ | independent of $[A]_0$ |
| 2 | $$\tau_{1/2} = \frac{1}{k[A]_0}$$ | inversely proportional to $[A]_0$ |

By measuring how $\tau_{1/2}$ changes with initial concentration, one can determine the order.

### 4. Isolation (Pseudo-Order) Method

When one reactant is in large excess, its concentration stays nearly constant.

Example:
$$
A + B \rightarrow P, \quad [B]_0 \gg [A]_0
$$

Then $[B] \approx [B]_0$ and rate law becomes:
$$
r = k [A]^m [B]_0^n = k' [A]^m
$$
where $k' = k[B]_0^n$ is a **pseudo rate constant**.

This converts the reaction to **pseudo–first order** (if $m = 1$), allowing easy analysis from $[A](t)$ data.

### 5. Graphical (Log–Log) Analysis

For any experiment:
$$
r = k [A]^m
$$
Taking logs:
$$
\log r = \log k + m \log [A]
$$
→ Plot $\log r$ vs. $\log [A]$: slope = m.

This is useful when direct integration is difficult.

### 6. Summary Table

| Method | Data Needed | Principle | Typical Use |
|--------|--------------|------------|--------------|
| Initial Rate | Initial rates at various concentrations | Slope of log–log plot | Fast, accurate for gas or solution kinetics |
| Integrated | Full $[A](t)$ data | Match linearized integrated law | Slow reactions, decomposition |
| Half-Life | Measure $\tau_{1/2}$ for various $[A]_0$ | $\tau_{1/2}$ dependence on $[A]_0$ | Simple decays |
| Isolation | Large excess of one reactant | Reduces multi-reactant kinetics to one-variable form | Bimolecular or complex systems |


## 1.3. Temperature Dependence of Reaction Rates: Arrhenius Equation and van’t Hoff’s Rule 

### A. Arrhenius Equation

The **rate constant** $k$ of most reactions increases with temperature.

**Empirical relation (Arrhenius, 1889):**

$$
\boxed{
k = A e^{-E_a / (RT)}
}
$$

where:

| Symbol | Meaning | Units |
|---------|----------|--------|
| $k$ | rate constant | depends on reaction order |
| $A$ | pre-exponential (frequency) factor | same as $k$ |
| $E_a$ | activation energy | J mol⁻¹ |
| $R$ | gas constant (8.314 J mol⁻¹ K⁻¹) |  |
| $T$ | absolute temperature (K) |  |

Taking logarithms:
$$
\boxed{
\ln k = \ln A - \frac{E_a}{R} \frac{1}{T}
}
$$

This gives a **linear relation** between $\ln k$ and $1/T$:

- **Slope:** $-E_a / R$  
- **Intercept:** $\ln A$

→ A straight line (Arrhenius plot) allows experimental determination of $E_a$.


**Two-temperature form**
If $k_1$ and $k_2$ are rate constants at $T_1$ and $T_2$:

$$
\boxed{
\ln\frac{k_2}{k_1} = \frac{E_a}{R}\left(\frac{1}{T_1} - \frac{1}{T_2}\right)
}
$$

This is often used for quick evaluation of activation energy without plotting.

**Physical meaning**
- $A$ - prefactor: frequency of effective collisions (and orientation probability)
- $E_a$ - (true) activation energy: minimum energy barrier that must be overcome for reaction


### B. Temperature Coefficient and van’t Hoff’s Rule

The **temperature coefficient** ($\gamma$) expresses how much the rate changes for a 10 °C rise in temperature.

**Definition**
$$
\gamma = \frac{k_{T+10}}{k_T}
$$

Typical values: $\gamma \approx 2\text{–}3$

That is, **the rate approximately doubles or triples** for each 10 °C increase in temperature (**van't Hoff's rule**).


From the Arrhenius equation:

$$
\ln \gamma = \frac{E_a}{R}\left(\frac{1}{T} - \frac{1}{T + 10}\right)
$$

If $E_a$ and $T$ are known, this relation allows computation of the temperature coefficient.

Example:
For $E_a = 50\,\text{kJ mol}^{-1}$, $T = 298\,\text{K}$:

$$
\ln \gamma = \frac{50\,000}{8.314}\left(\frac{1}{298} - \frac{1}{308}\right) = 0.65
$$

Hence:
$$
\gamma = e^{0.65} \approx 1.9
$$
→ The rate nearly doubles per 10 °C.


**Interpretation:**
- **High $E_A$** → strong temperature dependence  
- **Low $E_A$** → weak temperature dependence  
- Reactions involving **radicals or diffusion** often have small $E_A$, while **bond-breaking** reactions have large $E_A$.


### C. Empirical activation energy

Sometimes rate data deviate slightly from simple Arrhenius form; then one uses:

$$
k = A\, T^n\, e^{-E_a/(RT)}
$$

Taking logs:
$$
\ln k = \ln A + n \ln T - \frac{E_a}{R}\frac{1}{T}
$$

In this general situation, the **empirical activation energy, $E_A$** is determined by:
$$
E_A = RT^2 \frac{d \ln k}{dT} = n RT + E_a
$$

## Problem 1

A sample of 100 g or radioactive uranium (U) decays according to the following scheme:

$ ^{239}U \xrightarrow{ \tau_{1/2}\ = 23.5\ min}\ ^{239}Np \xrightarrow{ \tau_{1/2}\ = 2.35\ day}\ ^{239}Pu $

The half-lives for isotopes are given for individual processes indicated (as if these reactions were simple 1-st order reactions).

Calculate the masses of neptunium (Np) and plutonium (Pu) after: 1) 20 min; 2) 20 days after the start of decay.

Determine the maximal mass of neptunium that can be obtained from this uranium sample

### Solution

We are going to use the following equations:

For the sequential first-order scheme
$$
A \xrightarrow{k_1} I \xrightarrow{k_2} P
$$
with \($[I]_0 =0,\; =[P]_0 $\), the concentrations are:

$$
[A](t) = [A]_0 e^{-k_1 t},\\
[I](t) = \frac{k_1}{k_2 - k_1}\,[A]_0\big(e^{-k_1 t} - e^{-k_2 t}\big),\\
[P](t) = [P]_0 + \frac{k_2\!\left(1 - e^{-k_1 t}\right) - k_1\!\left(1 - e^{-k_2 t}\right)}{k_2 - k_1}\,[A]_0.
$$

This is equivalent to the (less compact) form:
$$
[P](t) = [P]_0 + \frac{k_1 k_2}{k_2 - k_1}[A]_0
\left(\frac{1 - e^{-k_1 t}}{k_1} - \frac{1 - e^{-k_2 t}}{k_2}\right).
$$

*Checks:* at \($t=0$\), \($P=P_0$\); as \($t\to\infty$\), \($P=P_0+[A]_0$\).

**Special case \($k_1 = k_2 = k$\):**
$$
[A](t) = [A]_0 e^{-k t},\\
[I](t) = [A]_0\,k t\,e^{-k t},\\
[P](t) = [P]_0 + [A]_0\Big(1 - e^{-k t}(1 + k t)\Big).
$$


```python
# From the half-lives we can compute the rate constants:
# tau = ln2 / k, so:

t1 = 23.5 # min
t2 = 2.35 * 24 * 60 # min


k1 = math.log(2)/t1 # min^-1
k2 = math.log(2)/t2 # min^-1

# Because the molar masses of all isotopes are the same, we can skip the conversion of masses 
# to moles and back, and hence we can use just the mass of U as [A]_0

# Case 1
t = 20 # min
m_U = 100 # g

m_Np = (k1/(k2-k1)) * m_U * (math.exp(-k1*t) - math.exp(-k2*t))
m_Pu = (k1*k2/(k2-k1)) * m_U * ( (1.0/k1)*(1.0 - math.exp(-k1*t)) - (1.0/k2)*(1.0 - math.exp(-k2*t)) )

print(F"m(Np) = {m_Np}, g;  m(Pu) = {m_Pu}, g")

# Case 2
t = 20 * 24 * 60 # days

m_Np = (k1/(k2-k1)) * m_U * (math.exp(-k1*t) - math.exp(-k2*t))
m_Pu = (k1*k2/(k2-k1)) * m_U * ( (1.0/k1)*(1.0 - math.exp(-k1*t)) - (1.0/k2)*(1.0 - math.exp(-k2*t)) )

print(F"m(Np) = {m_Np}, g;  m(Pu) = {m_Pu}, g")


# The time at which the maximal amount of the intermediate is formed is
t_max = math.log(k2/k1)/(k2-k1)

# Evaluate the amount of Np at this time:
t = t_max

m_Np = (k1/(k2-k1)) * m_U * (math.exp(-k1*t) - math.exp(-k2*t))

print(F"m(Np)_max = {m_Np}, g")
```

### Answers

* Case 1: m(Np) = 44.46 g; m(Pu) = 0.1 g
* Case 2: m(Np) = 0.276 g; m(Pu) = 99.72 g
* Max mass of Np is 96.58 g

## Problem 2

For a first-order reversible reaction

$$
A \;\underset{k_2}{\stackrel{k_1}{\rightleftharpoons}}\; B
$$

equilibrium constant is K = 10 and the forward reaction rate constant is $k_1 = 0.2 s^{-1}$. Calculate the time at which the concentrations of substances A and B become equal, if the initial concentration of substance B is 0.

### Solution

$[A] = [A]_0 - x = [B]_0 + x$

Considering that $[B]_0$, we obtain the condition at which the concentrations of substances A and B become equal:  $x = 0.5 [A]_0$ 

but we also know that 

$x = x_{eq} (1 - exp(- (k_1 + k_2)t ) ) $

For initial zero concentration of B, we get:

$x_{eq} = \frac{k_1 [A]_0  - k_2 [B]_0 }{k_1 + k_2} = \frac{k_1 [A]_0  }{k_1 + k_2}$

So we obtain:

$0.5 [A]_0 = x = \frac{k_1 [A]_0  }{k_1 + k_2} (1 - exp(- (k_1 + k_2)t ) )$

or 

$0.5 = \frac{k_1 }{k_1 + k_2} (1 - exp(- (k_1 + k_2)t ) )$


We already know $k_1$. We can determine $k_2$ from $K = \frac{k_1}{k_2}$

```python
K = 10
k1 = 0.2 # s^-1
k2 = k1/K

t = - math.log(1 - 0.5*(k1+k2)/k1) / (k1+k2)

print(F"time = {t} s")
```

## Problem 3

The kinetics of the $2A \to B$ reaction occuring in the liquid phase was studied spectrophotometrically and the following data were obtained:


| t, min               | 0   |  10   |  20  |  30   |   40 | infinity  |
| --------------------- | --- | ----- | ---- | ----- | ---- | --------- | 
| [B], mol L^{-1} | 0.0 | 0.089 | 0.153 | 0.200 | 0.230 | 0.312  |


Determine the reaction order and rate constant


### Solution

```python
import math
import matplotlib.pyplot as plt
import numpy as np

t = [0, 10, 20, 30, 40, 1e10]
B = [0, 0.089, 0.153, 0.200, 0.230, 0.312]

# First, let's convert these data to concentrations of A
# The initial concentration of A is the twice of the final concentration of B
A0 = 2.0*B[5]

# The concentrations of the A are thus:
A = []
for i in range(6):
    A.append( A0 - 2*B[i])
    
print("A = ", A)
print("B = ", B)

# Now, let's compute the rates at different times, by doing the finite difference calculations, except for the last pair of points
rates = []
for i in range(4):
    r = -  (A[i+1] - A[i]) /( t[i+1] - t[i])
    rates.append( r )
    
print(rates)   
```

```python
# Let's also plot rates vs initial concentrations 
plt.plot(rates, A[0:4])
```

This plot shows that the rate is a linear function of the concentration. The slope in these coordinates would give the rate constnat

```python
y = rates
x = A[0:4]

print(x, y)

ave = 0.0
for i in range(3):
    # In principle, this should work 
    #slp = y[i]/x[i]
    # this is a slightly better
    slp = (y[i+1]-y[i])/(x[i+1]-x[i])
        
    ave = ave + slp 
    print(F"Slope {i} = {slp} ")
    
ave = ave / 3
print(F"rate constant = {ave}" )
```

```python
# Now in a logarithmic scale:
plt.plot(np.log(np.array(rates)), np.log(np.array(A[0:4])) )
```

The slope in these coordinates would give us the order wrt to this component. Visually, it is close to 1, confirming the previos observation of the 1-st order reaction data.

Now compute the average value of the slope in these coordinates:

```python
# The slope would give use the order wrt component A
x = np.log(np.array(rates))
y = np.log(np.array(A[0:4]))

print(x, y)

for i in range(3):
    slp = (y[i+1] - y[i])/(x[i+1] - x[i])
    print(F"Slope {i} = {slp} ")
```

```python
# Cheaters solution:
from scipy.optimize import curve_fit


t = np.array([0, 10, 20, 30, 40])
B = np.array([0, 0.089, 0.153, 0.200, 0.230])

plt.plot(t, B)

def exp_fit(t, k):
    return 0.312 * (1 - np.exp(-k*t))

[k], _ = curve_fit( exp_fit, t, B, bounds=([0.0],[np.inf]))

plt.plot(t, exp_fit(t, k), ls='-',  color="black", label="fit", lw=5)

print(k)
```

### Answer

* First order wrt A
* k = 0.034 $min^{-1}$ (exact fitting), or 0.03 $min^{-1}$ ("manual" solution) or anything slightly less accurate is fine

## Problem 4

In a first-order reaction the conversion degree is 30% at 30 min mark at temperature of 25 Celsius. At what temperature would the conversion degree be 60% at 40 min mark, if the activation energy is 30 kJ/mol.

### Solution

For the first-order reaction, we have $[A] = [A]_0 exp(-kt)$. A 30% degree of conversion means that at that time $[A] = 0.7[A]_0$. Likewise, a 60% degree of conversion means that $[A] = 0.4[A]_0$.

```python
# First, compute the rate constant at 25 C:
# 0.7 = exp(-k_25*30) => k_25 = -ln(0.7)/30  # min^-1
k_25 = -np.log(0.7)/30.0
print(F" rate constant at 25 C = {k_25} min^-1")

# Second, compute the rate constant at the unknown temperature:
# 0.4 = exp(-k_T*40) => k_T = -ln(0.4)/40  # min^-1
k_T = -np.log(0.4)/40.0
print(F" rate constant at T C = {k_T} min^-1")

# from k = A * exp(-E_a/RT) we have:
# ln(k_T/k_25) = -(E_a/R) * ( 1/T - 1/T_25 )
# where T_25 = 25 + 273.15

T_25 = 25 + 273.15
R = 8.314 # J/mol*K
E_a = 30000 # J/mol

invT =  - (R/E_a) * np.log(k_T/k_25) + 1/T_25
T = 1.0/invT
T = T - 273.15
print(F"Temperature = {T} Celsius")
```

## Problem 5

For some first-order reaction, the following values of the reaction rate constant at different temperatures are determined:

| Temperature, Celsius | Rate constant, $s^{-1}$ |
| ---- | --- |
| 0  | 2.46 $\times 10^{-5}$ |
| 20 | 47.5 $\times 10^{-5}$ |
| 40 | 576 $\times 10^{-5}$ |
| 60 | 5480 $\times 10^{-5}$ |

Compute the activation energy and pre-exponential factor of the rate constant. Compute the half-life time at 25 Celsius.

### Solution

```python
# Initial data

T = np.array([0.0, 20.0, 40.0, 60.0]) + 273.15
print(T)

k = np.array([2.46, 47.5, 576, 5480]) * 1e-5
print(k)
```

```python
# Do Arrhenius plot:
invT = 1.0/T
print(invT)

lnk = np.log(k)
print(lnk)

plt.plot(invT, lnk)
```

```python
# Now fit the Arrhenius plot to linear function
def lin_fit(x, a, b):
    return a + b * x

[A, B], _ = curve_fit( lin_fit, invT, lnk, bounds=([-np.inf, -np.inf],[np.inf, np.inf]))

plt.plot(invT, lnk)
plt.plot(invT, lin_fit(invT, A, B))

# The fitting coefficients are:
# A = ln(pre_exponent)
# B = - E_a/R

pre_exponent = np.exp(A)
R = 8.314 # J/mol*K
E_a = -B * R

print(F"pre_exponent = {pre_exponent:.2e}, E_a = {E_a} J/mol")
```

```python
# Now we can use the the Arrhenius formula together with the computed parameters to estimate 
# the rate constant at a different temperature
T_new = 25 + 273.15
k = pre_exponent * np.exp(-E_a/(R*T_new) )

print(F"rate constant at 25 Celsius = {k:.2e}")

# and compute the half-life:
tau = np.log(2)/k

print(F"half-time is {tau} seconds")
```

### Answers

* pre_exponent = 8.95e+13

* activation energy = 97 kJ/mol

* half-life = 772 s
